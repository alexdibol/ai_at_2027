# NB05 — Governed Meta-Agent

## Introduction

NB05 introduces the planning and orchestration layer of the autonomous research architecture. The earlier notebooks deliberately built this capability from the bottom up. NB00A established the system protocol and its research-only constitution. NB00 created a reproducible synthetic equity market. NB00B demonstrated a complete KNN research slice from data through stress analysis. NB01 expanded that slice into a comparative laboratory of models, strategies, and portfolios. NB02 added execution simulation, independent risk, governance gates, provenance, and audit. NB03 transformed validated functions into 35 tools, nine skills, and 13 bounded agent roles. NB04 then organized those roles into four governed constellations with explicit state ownership, dissent, budgets, handoffs, and human checkpoints. NB05 asks the next unavoidable question: who interprets a research mission, decomposes it, discovers the required capabilities, routes work across those constellations, monitors the mission, and stops it when policy requires?

The answer is the governed meta-agent. A meta-agent is not a super-agent, a chief executive, or an unrestricted intelligence standing above the institution. It is a bounded orchestration role whose purpose is to transform a semantically sufficient research objective into an observable dependency plan and route each task only through capabilities already validated by the system. It coordinates work; it does not inherit the union of every agent’s permissions. It may discover, decompose, route, allocate, aggregate, detect, request, retry, degrade, and terminate. It may not invent a tool, change the human objective, suppress dissent, approve its own recommendation, override risk, expand its budget, delegate around a denial, claim unavailable evidence, connect to a broker, execute a live trade, or authorize deployment.

This distinction is central to institutional AI. As systems acquire more specialized agents, merely adding a planner can quietly create a dangerous privilege-escalation path. An orchestrator may appear to need broad access because it touches every stage of the workflow. Yet broad visibility is not the same as broad authority. NB05 therefore treats the meta-agent as a controller of sequence and evidence rather than as the owner of the underlying powers. Every proposed delegation remains subject to the same effective-authority intersection used in NB03 and NB04: system policy, mission scope, registered route, agent permission, skill allowlist, tool contract, remaining budget, and approval state must all permit the next action. A denial at any layer ends that route.

The notebook begins by encoding the meta-agent’s specification in machine-readable form. Its working memory is limited to one mission and is cleared at closure. Durable reads are restricted to validated registries, policies, and immutable audit evidence. Durable writes are limited to append-only audit events, approved artifacts, and unresolved questions. Private credentials, unapproved personal data, and hidden chain-of-thought records are prohibited. These rules define memory as a governed institutional resource. The planner receives enough context to coordinate the mission without becoming a repository for information that it is neither entitled nor required to retain.

Resource limits are equally explicit. The meta-agent may construct at most 12 plan steps, issue at most eight delegations, make at most 48 tool calls, retry a failed task once, and consume no more than 100 cost units. These constraints do more than manage expense. They limit looping, recursive delegation, repeated tool use, and the tendency to continue indefinitely when evidence remains incomplete. When a budget is exhausted, the correct behavior is to terminate, preserve partial evidence and open questions, and return control to the accountable human owner. The meta-agent is prohibited from increasing its own allowance.

Capability discovery precedes planning. The notebook verifies the inherited architecture of 13 agents, four constellations, nine skills, and 35 tools. Every enumerated agent, constellation, and skill is validated and carries no live authority. The discovery record includes a capability hash so that the mission can be tied to a specific registry state. This prevents the planner from treating a plausible description as an available capability. If the required capability is absent, the mission is denied or escalated; the system does not fabricate a substitute simply to appear helpful.

Once a mission is admitted, NB05 turns it into a five-task dependency graph. The Research Team produces the evidence bundle. The Strategy Team receives that evidence and produces an experimental strategy proposal. The Portfolio Team converts the proposal into a constrained simulated portfolio. The Risk Governance Team independently challenges the research, strategy, and portfolio evidence. The same control constellation then assembles the final audit bundle and preserves unresolved questions. This sequence embodies separation of duties. Risk depends on the upstream evidence but is not owned by the upstream producers, and the audit task depends on completion of the independent challenge.

Planning does not itself authorize delegation. Before every route, the meta-agent must preserve the synthetic-equity scope, research-only status, registered ownership, absence of live execution, and budget compliance. The notebook’s trace records the mission identifier, step, task, route, decision, tool-call count, cost, cumulative cost, outcome, and provenance hash. The valid mission completes five delegated tasks, consumes 95 of 100 available cost units, and stays within the tool-call ceiling. The trace makes orchestration reviewable: a later auditor can reconstruct not only what was produced, but why each task was routed, which resources it consumed, and which evidence lineage accompanied it.

Evidence aggregation is intentionally conservative. The meta-agent does not average away disagreement or select the most convenient claim. The notebook preserves a conflict about economic persistence: one evidence item describes the signal as promising, while another describes it as uncertain. That conflict remains OPEN and changes the governance state. The meta-agent may detect and package the disagreement, but an unresolved material conflict is a termination or escalation condition rather than an invitation to manufacture consensus. This preserves the institutional value of independent challenge.

Recovery is likewise bounded. One retry is permitted for a failed delegated task. If the failure remains recoverable, the task may be reassigned to an approved route with new provenance while unaffected tasks continue. Recovery cannot erase the original failure, silently reset cost, or broaden permission. A successful recovery means that the system followed its authorized recovery contract; it does not mean that the failure never happened.

The adversarial laboratory tests ten mission types. A valid objective completes as research-only. An ambiguous objective escalates for missing semantics. Deceptive intent is denied. Contradictory authority constraints block the mission. An unavailable capability and a live-execution request are denied. A material evidence conflict escalates. Budget exhaustion terminates the mission with partial evidence. A delegate failure recovers through one retry and approved reassignment. Human denial terminates the mission. These outcomes establish a crucial standard: completion is only one successful governance response. Refusal, blocking, escalation, controlled recovery, and termination are equally legitimate when they match policy.

NB05 therefore validates orchestration rather than investment performance. It does not prove that the proposed strategy is profitable, that language-model planning is reliable in open environments, or that the system is ready for production. The evidence is deterministic, synthetic, and educational. The notebook demonstrates that a meta-level planner can be made bounded, inspectable, resource-aware, conflict-sensitive, and subordinate to human authority. It deliberately leaves autonomous integration pending.

Within the course progression, NB05 is the bridge from governed organization to integrated autonomy. NB04 supplied the constellations and their institutional constitution. NB05 supplies the planner that can interpret an admitted mission and coordinate those teams without rewriting their permissions. NB06 will integrate the components into an autonomous-system release candidate with state-machine behavior, checkpointing, monitoring, and full-mission validation. NB07 will then build a controlled research-to-LEAN experiment bridge. The doctrine remains unchanged: increasing orchestration may expand analytical reach, but it must never create authority that was not explicitly granted.

## 4. Decomposition and routing

A sufficient objective becomes a five-task directed dependency graph. Research produces evidence; strategy converts evidence into an experimental proposal; portfolio construction converts the proposal into bounded simulated positions; risk and governance independently challenge all upstream outputs; and audit closes the mission. Explicit owners and dependencies prevent silent role substitution and premature downstream action.

```python
PLAN=json.loads("[{\"task_id\": \"T1\", \"capability\": \"research\", \"depends_on\": [], \"owner\": \"research_team\", \"budget\": 20, \"output\": \"evidence bundle\"}, {\"task_id\": \"T2\", \"capability\": \"strategy\", \"depends_on\": [\"T1\"], \"owner\": \"strategy_team\", \"budget\": 20, \"output\": \"experimental strategy proposal\"}, {\"task_id\": \"T3\", \"capability\": \"portfolio\", \"depends_on\": [\"T2\"], \"owner\": \"portfolio_team\", \"budget\": 20, \"output\": \"constrained simulated portfolio\"}, {\"task_id\": \"T4\", \"capability\": \"risk\", \"depends_on\": [\"T1\", \"T2\", \"T3\"], \"owner\": \"risk_governance_team\", \"budget\": 25, \"output\": \"independent challenge and audit\"}, {\"task_id\": \"T5\", \"capability\": \"audit\", \"depends_on\": [\"T4\"], \"owner\": \"risk_governance_team\", \"budget\": 10, \"output\": \"final audit bundle and unresolved questions\"}]")
assert len(PLAN)==5 and PLAN[-1]['depends_on']==['T4']
print('5-task dependency plan PASS')

task_ids = [task['task_id'] for task in PLAN]
assert len(task_ids) == len(set(task_ids)) == 5
known = set()
for task in PLAN:
    assert set(task['depends_on']).issubset(known)
    known.add(task['task_id'])
registered_constellations = {item['id'] for item in DISCOVERY['items'] if item['kind'] == 'constellation'}
assert all(task['owner'] in registered_constellations for task in PLAN)
assert sum(task['budget'] for task in PLAN) == 95
print('acyclic plan, topological order, registered owners, and cost reservation PASS')
```

## Frequently Asked Questions — NB05

## 1. What is the governed meta-agent?

The governed meta-agent is the system’s bounded planning and orchestration role. It receives a semantically sufficient synthetic-equity research objective, discovers validated capabilities, decomposes the mission, builds dependencies, assigns bounded resources, routes tasks, aggregates evidence, detects conflicts, requests approval when necessary, and records the mission trace. It coordinates institutional work but cannot approve, deploy, or trade.

## 2. Is the meta-agent a more powerful general-purpose agent?

No. It has broader visibility over the workflow but not broader substantive authority. It does not inherit the combined permissions of the agents and constellations it coordinates. Every delegated action remains constrained by system policy, mission scope, the selected route, the acting agent’s permissions, the skill allowlist, the tool contract, the available budget, and the current approval state.

## 3. Why is a meta-agent needed after NB04 already created constellations?

NB04 defines how four institutional teams cooperate once work reaches them. NB05 supplies the mission-level controller that interprets an admitted objective, decides which team should act when, expresses dependencies, allocates resources, follows handoffs, aggregates results, and determines whether the mission should continue or stop. Constellations coordinate roles; the meta-agent coordinates the mission across constellations.

## 4. What makes an objective semantically sufficient?

A sufficient objective clearly identifies the research question, artifact or target, synthetic-equity scope, evidence boundary, requested output, constraints, and accountable decision owner. The planner must understand what is being requested before evaluating whether it is allowed. Ambiguity does not produce a guessed plan. It produces escalation and a request for the missing fields.

## 5. Which actions may the meta-agent perform?

Its allowlist includes discovering registered capabilities, decomposing research tasks, building a dependency plan, routing to a registered agent or constellation, allocating bounded budgets, aggregating evidence, detecting conflicts, requesting approval, retrying once, degrading gracefully, and terminating. These actions concern orchestration; they do not include approval, deployment, brokerage connectivity, or live execution.

## 6. Which behaviors are categorically prohibited?

The meta-agent may not invent capabilities, alter the human objective, hide or overwrite dissent, self-approve, override risk, enlarge its own budget, delegate around a denial, claim evidence that is unavailable, connect to a brokerage, execute live trades, or authorize deployment. Urgency, confidence, or unanimous agent agreement does not suspend these prohibitions.

## 7. How does capability discovery work?

The planner reads the validated registries inherited from NB03 and NB04. The discovery snapshot reports 13 agents, four constellations, nine skills, and 35 tools, and includes an integrity result and capability hash. An object is usable only through a registered and permitted route. A plausible capability name is not enough, and an absent capability causes denial or escalation.

## 8. Why does the discovery record include a hash?

The capability hash binds the plan to a specific registry state. Without it, later reviewers could not establish whether the planner used the same set of roles and capabilities that had been validated when the mission began. A production system would also sign and version registries, enforce access controls, and monitor unauthorized changes.

## 9. How is the mission decomposed?

NB05 builds five tasks. T1 assigns research evidence to the Research Team. T2 assigns the experimental strategy proposal to the Strategy Team and depends on T1. T3 assigns the constrained simulated portfolio to the Portfolio Team and depends on T2. T4 assigns independent challenge to the Risk Governance Team and depends on T1 through T3. T5 creates the final audit bundle after T4.

## 10. Why is the plan represented as a dependency graph?

The graph makes sequencing constraints explicit. A strategy cannot legitimately precede its evidence, a portfolio cannot precede the strategy specification, and the final audit cannot precede independent challenge. Explicit dependencies also enable cycle detection, bounded parallelism, topological validation, and reconstruction of the reasoning process without relying on an informal narrative of what happened.

## 11. How does the meta-agent choose a route?

It matches the task’s declared capability and output contract with a validated agent or constellation. It then checks scope, registration, evidence, permissions, tool exposure, budget, and approval requirements. The effective route is the intersection of these constraints. If any required element is missing or denies the action, the planner must stop that route rather than delegate creatively around the restriction.

## 12. Can the meta-agent invoke any tool directly?

No. Requests do not jump from the planner to arbitrary tools. A tool must be registered, exposed through a validated skill, available to the acting agent, permitted by the constellation and system policy, compatible with the mission scope, and within budget. The meta-agent coordinates the route but does not bypass the tool–skill–agent authorization structure established in NB03.

## 13. What budgets constrain the mission?

The meta-agent is limited to 12 plan steps, eight delegations, 48 tool calls, one retry per task, and 100 cost units. The five-task plan reserves 95 units, while the recorded trace remains inside the tool-call ceiling. These budgets bound cost, recursion, repeated failure, and indefinite analysis. The meta-agent cannot increase them itself.

## 14. What happens when a budget is exhausted?

The mission terminates with reason BUDGET_EXHAUSTED. The system returns the evidence already obtained, unresolved questions, remaining dependencies, and an auditable cost record. Termination is not treated as an analytical failure to conceal. Continuing would require a new authorization or mission, not an undocumented increase by the planner.

## 15. How is memory governed?

Working memory is limited to one mission and is cleared at closure. Durable reads include validated registries, policies, and immutable audit evidence. Durable writes include append-only audit events, approved artifacts, and unresolved questions. Private credentials, unapproved personal data, and hidden chain-of-thought content are forbidden. Memory access is therefore tied to institutional purpose and accountability.

## 16. What does “policy before delegation” mean?

Planning a task does not authorize it. Before dispatching each task, the meta-agent must confirm that the mission remains within synthetic equities, research-only, free of live execution, assigned to a registered route, and within the available budget. The action is delegated only after these conditions pass, and the same controls are reconsidered as mission state changes.

## 17. How are evidence conflicts handled?

Conflicting claims remain separate and visible. In the notebook, economic persistence is described as both promising and uncertain, so the aggregate status becomes CONFLICT and the disagreement remains OPEN. The planner may package both claims and request independent review, but it may not average the conflict away, select the convenient claim, or silently rewrite an upstream conclusion.

## 18. Which decisions require human approval?

Human approval is required for a material objective change, strategy promotion, a risk override or limit change, an external-platform write, deployment, or live action. The meta-agent may assemble the relevant evidence, dissent, assumptions, and requested authority, but it cannot become the approver. A human denial is final for the mission and produces controlled termination.

## 19. When must the mission terminate?

Termination conditions include a prohibited objective, failed semantic sufficiency, an absent required capability, exhausted budget, unresolved material conflict, human denial, and mission completion. Completion is itself a termination condition because working memory must close cleanly. The planner is not permitted to invent new work merely to remain active.

## 20. How does bounded recovery work?

One retry is permitted for each failed task. If policy allows, the meta-agent may then reassign the task to an approved alternate route, append new provenance, and continue unaffected tasks. The original failure remains in the record. Recovery may not reset the budget, erase dissent, broaden scope, or route around a denial.

## 21. What does the trace record?

Each trace event contains the mission identifier, step number, task identifier, route, decision, tool-call count, cost units, cumulative cost, outcome, and provenance hash. The five-step valid trace reaches a cumulative cost of 95 units. This record supports reconstruction, budget verification, route review, and comparison between the approved plan and actual delegated behavior.

## 22. Which adversarial cases are tested?

The laboratory includes a valid objective, ambiguous objective, deceptive objective, contradictory objective, unachievable objective, live-execution request, material evidence conflict, budget exhaustion, delegate failure, and human denial. Each case has an expected governed status, reason, recovery instruction when applicable, false live-authority flag, pass indicator, and 64-character hash.

## 23. What do DENIED, BLOCKED, ESCALATED, TERMINATED, and RECOVERED mean?

DENIED means the requested capability or authority is categorically unavailable. BLOCKED means contradictory constraints prevent progress until clarified. ESCALATED means accountable judgment or missing semantics are required. TERMINATED means a hard boundary or human denial ends the mission. RECOVERED means the system followed its authorized retry and reassignment procedure without hiding the failure.

## 24. What is the result of the valid mission?

The valid objective reaches COMPLETED_RESEARCH_ONLY. All five delegated tasks complete, the final trace consumes 95 cost units, and the planner produces an evidence-bearing audit package. This status does not promote a strategy, authorize production, or create a live order. It states only that the governed research workflow completed within its declared boundary.

## 25. Why is live authority false in every result?

Live authority is not a reward for analytical success. It is absent by design from the meta-agent, every discovered agent, every constellation, and all ten test outcomes. The live-execution case is explicitly denied. A research artifact may later support a human decision, but evidence cannot manufacture permissions that the architecture never granted.

## 26. What does NB05 validate, and what does it not prove?

NB05 validates deterministic orchestration specifications, registry discovery, a dependency plan, budget checks, policy gates, evidence-conflict preservation, trace coherence, and ten expected governance outcomes in a synthetic research environment. It does not prove investment profitability, open-world planning reliability, production security, legal compliance, broker readiness, or resilience to real-time market and infrastructure failures.

## 27. What does NB05 hand to NB06?

NB06 receives the validated meta-agent specification, registry snapshot and capability hash, five-task plan, budget and policy rules, evidence-conflict state, provenance-bearing execution trace, adversarial scenario results, termination conditions, and approval boundaries. The next notebook integrates these components into a release candidate with explicit states, checkpoints, monitoring, and complete mission execution.

## 28. What is the final lesson of NB05?

Orchestration must be governed as rigorously as execution. A planner creates institutional value when it can translate objectives into coherent work, preserve dependencies and dissent, allocate scarce resources, and stop safely. It becomes dangerous when coordination is mistaken for authority. The meta-agent may organize intelligence across the system; humans retain accountability for material decisions and external effects.

## Why NB05 Is Not Yet an Autonomous Architecture

NB05 is an important architectural milestone, but it is not yet an autonomous system. It defines how a governed meta-agent should behave and validates a predetermined example of planning and routing. The notebook provides the constitution, vocabulary, limits, and expected outcomes of autonomous orchestration—but it does not yet perform that orchestration dynamically.

The clearest description is:

> **NB05 is a validated blueprint and test harness for autonomy, not yet an operating autonomous architecture.**

## What NB05 already accomplishes

NB05 establishes several essential foundations. It defines the meta-agent’s permitted actions, prohibited behaviors, memory boundaries, budgets, termination conditions, approval requirements, and research-only status. It also specifies the institutional progression from research through strategy design, portfolio construction, independent risk review, and audit.

The notebook verifies that:

- Only registered capabilities may be used.
- The plan respects its dependency structure.
- Budgets remain within predetermined limits.
- Evidence conflicts remain visible.
- Human approval is required for material decisions.
- Live execution remains prohibited.
- Ambiguity, deception, failure, budget exhaustion, and human denial produce the expected governed outcomes.

These are necessary properties of an autonomous architecture. They are not, however, sufficient to make the architecture autonomous.

## Why the current implementation is not autonomous

The fundamental limitation is that the important decisions have already been made before the notebook runs.

The mission is already decomposed into five tasks. Each task already has a capability label, an owner, a budget, a dependency structure, and an expected output:

| Task | Predetermined label | Predetermined route |
|---|---|---|
| T1 | Research | Research Team |
| T2 | Strategy | Strategy Team |
| T3 | Portfolio | Portfolio Team |
| T4 | Risk | Risk Governance Team |
| T5 | Audit | Risk Governance Team |

The meta-agent does not generate this plan. It loads it.

Similarly, capability discovery is represented by a static registry snapshot. Policy decisions are represented by predefined Boolean values. The execution trace is loaded as a completed record, and the adversarial outcomes are supplied with their expected decisions already attached.

Consequently, the notebook answers:

> **Is this predetermined plan internally coherent and consistent with policy?**

It does not yet answer:

> **Can the system receive a new objective, understand it, construct a plan, classify its tasks, select authorized routes, execute the plan, respond to new evidence, recover from failures, and close the mission without having those decisions prepared in advance?**

That second question is the test of operational autonomy.

## What is missing

A genuinely autonomous research architecture needs an executable closed loop:

$
\text{Objective}
\rightarrow
\text{Interpret}
\rightarrow
\text{Decompose}
\rightarrow
\text{Label}
\rightarrow
\text{Route}
\rightarrow
\text{Execute}
\rightarrow
\text{Observe}
\rightarrow
\text{Evaluate}
\rightarrow
\text{Replan or Stop}
$

NB05 currently represents most of these stages as static objects. It does not yet contain the engines that create them at runtime.

### Dynamic objective interpretation

The system must receive a new human objective and determine whether it contains sufficient information. It should identify the scope, expected output, evidence boundary, constraints, decision owner, and prohibited effects.

An incomplete objective should trigger clarification. A prohibited objective should be denied. These decisions must be computed from the request rather than retrieved from a predefined test record.

### Dynamic task decomposition

The system must convert the admitted objective into tasks. The number and nature of those tasks should depend on the objective.

A data-quality investigation, for example, should not automatically produce the same five-task plan as a strategy-evaluation mission. The current plan is appropriate for its example, but it should be an output of the planning engine rather than a constant embedded in the notebook.

### Task labeling

Each decomposed task must be classified against a controlled ontology:

- Data research
- Feature engineering
- Model development
- Strategy design
- Backtesting
- Portfolio construction
- Execution simulation
- Risk review
- Governance review
- Audit
- Research-to-LEAN conversion

The labeler determines what the task requires. It should not decide who performs it.

### Dynamic route selection

The router must evaluate the registry and calculate which agent or constellation is eligible.

The selection should depend on:

- Required capabilities
- Scope permission
- Evidence availability
- Input and output compatibility
- Separation of duties
- Remaining budget
- Current workload
- Independence requirements
- Approval state
- Prohibited side effects

The route should therefore be computed and explained:

    {
        "task_id": "T4",
        "selected_route": "risk_governance_team",
        "reason": "complete capability coverage and required independence",
        "rejected_candidates": {
            "portfolio_team": "insufficient independence",
            "risk_officer": "incomplete governance coverage"
        }
    }

### Runtime execution and state transitions

The system must invoke the selected capability, receive its actual output, validate that output, update the shared mission state, and determine the next permitted action.

This requires an explicit state machine, for example:

$
\text{RECEIVED}
\rightarrow
\text{ADMITTED}
\rightarrow
\text{PLANNED}
\rightarrow
\text{ROUTED}
\rightarrow
\text{EXECUTING}
\rightarrow
\text{CHALLENGED}
\rightarrow
\text{AUDITED}
\rightarrow
\text{CLOSED}
$

Alternative states must include:

$
\text{BLOCKED},\quad
\text{ESCALATED},\quad
\text{DENIED},\quad
\text{RECOVERING},\quad
\text{TERMINATED}
$

### Feedback and replanning

An autonomous system cannot assume that every task will succeed exactly as planned. It must compare actual outputs with expected output contracts.

If evidence is missing, a tool fails, a conflict appears, or a budget becomes insufficient, the system must decide whether to retry, reassign, revise the remaining plan, escalate, or terminate.

The current NB05 contains examples of these outcomes, but the outcomes are prerecorded. An autonomous system must produce them from runtime conditions.

This limitation does not make NB05 unnecessary or incorrectly designed. It performs the indispensable step of defining the governance contract before activating autonomy.

Attempting dynamic planning and delegation before establishing permissions, budgets, denial rules, human checkpoints, provenance requirements, and termination conditions would create an ungoverned system.
