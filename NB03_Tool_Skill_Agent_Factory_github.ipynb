# NB03 — Tool Factory, Skill Library, and Bounded Agents

## Introduction

NB03 marks a decisive change in the architecture of the course. The earlier notebooks constructed and evaluated a quantitative research process: NB00A defined the operating protocol, NB00 created the synthetic market, NB00B demonstrated a minimal KNN vertical slice, NB01 expanded the laboratory across models, signals, strategies, and portfolios, and NB02 placed the selected proposal behind execution, independent-risk, governance, provenance, and audit controls. Those notebooks produced useful functions and evidence, but functions scattered across notebooks are not yet an autonomous institution. NB03 therefore asks a new engineering question: how do we transform validated research operations into reusable capabilities that an agent can discover and invoke without receiving unlimited authority?

The answer is a capability factory organized into three layers: tools, skills, and agents. A tool is an atomic, machine-invocable capability with a declared purpose, inputs, outputs, error conditions, permissions, provenance, tests, version, and lifecycle state. A skill is a repeatable procedure that composes approved tools under evidence requirements, quality checks, boundaries, escalation rules, and stopping conditions. An agent is a bounded institutional role that may use only an explicit set of skills and tools, within a limited memory scope, step budget, call budget, permission envelope, policy gates, and handoff contract. These distinctions prevent the word “agent” from becoming a vague synonym for software. They make capability, procedure, responsibility, and authority separately inspectable.

The notebook registers 35 tools inherited from the validated work of Days 1 through 5. They cover specification validation, synthetic-data generation, canonical database construction, dataset integrity, point-in-time loading and feature engineering, chronological partitioning, KNN and common-model training, signal construction, strategy evaluation, portfolio constraints, backtesting, execution simulation, order reconciliation, market-impact estimation, independent risk, stress testing, limits, semantic sufficiency, action authorization, provenance verification, and audit logging. Registration does not merely place a name in a catalog. Every tool specification states what the tool is for, which evidence must enter, what output contract must be satisfied, how failure is represented, which effects are allowed, which effects are denied, where the tool came from, and how it was tested.

Permissions are especially important. The registered tools may read approved research artifacts and calculate within the synthetic-equity environment. They explicitly deny execution in live markets and brokerage systems. Side effects are declared rather than assumed. Each tool carries golden-path and denial-path tests, and each requires content-hash provenance. The status `validated` means the specification has passed the notebook’s admission checks; it does not mean the tool is universally correct, commercially deployable, or permitted to trade. In this course, registration is evidence-bounded. No claim may exceed the environment, tests, and authority actually represented by the specification.

The second layer contains nine skills: Data Research, Model Development, Strategy Design, Backtesting, Portfolio Construction, Risk Review, Governance Review, Audit Bundle, and Research to LEAN. A skill is more than a list of functions. It defines an objective and procedure, names the tools that may be used, states the evidence required before work begins, identifies quality controls, declares boundaries, and specifies when to stop or escalate. For example, the Data Research skill requires a request identifier, an approved synthetic-equity input, and content hashes. It checks schemas, point-in-time integrity, and provenance. It must stop when evidence is incomplete, permission is denied, its budget is exhausted, or a required tool is unregistered.

The third layer registers 13 agents representing recognizable institutional roles: Research Analyst, Data Scientist, Experiment Designer, Strategy Designer, Portfolio Constructor, Risk Officer, Governance Critic, Audit Reviewer, Execution Simulation Agent, QuantConnect Code Generator, Code Report Generator, Governance Strategy Evaluation Agent, and Research-to-QuantConnect Converter. Every agent receives an allowlist derived from its approved skills. The role can read governed evidence, produce draft research artifacts, calculate within synthetic equities, and simulate in a sandbox. It cannot approve its own work, promote a strategy, override risk, connect to a broker, or execute in live markets. Memory is limited to the current task envelope, durable access is read-only, and the standard budget permits at most eight steps and 12 tool calls.

The policy gates make this boundedness operational. Before an agent can act, the request must be semantically sufficient, the proposed capability must be registered, the evidence must be complete, budget must remain available, separation of duties must be respected, and the action must remain research-only. Risk-bearing recommendations are handed to the Risk Officer. Approval requests, exceptions, ambiguity, prohibited actions, and other material decisions are handed to the human governance owner with the request, evidence, denial reason, and audit hash. The system is not designed to make every request succeed. A blocked proposal is a valid system result when the requested action exceeds the role’s mandate or the available evidence.

NB03 tests this architecture with seven scenarios. A valid research request completes after passing the policy gates. Six adversarial requests are correctly denied: an attempt to invoke a hallucinated or unregistered capability, agent overreach beyond the allowlist, incomplete evidence, budget exhaustion, unauthorized live execution, and self-approval by a strategy role. Each result records the responsible agent, status, reasons, absence of live authority, expected outcome, pass flag, and audit hash. The tests therefore measure both productive capability and disciplined refusal. An autonomous research system must know not only how to perform work, but also when it lacks the authority, evidence, budget, or registered capability required to continue.

Within the course progression, NB03 begins the organizational layer of the machine. Days 1 through 5 built the world, the quantitative intelligence, and its controls. Day 6 converts that foundation into a reusable capability architecture. Day 7 will organize these bounded roles into Research, Strategy, Portfolio, and Risk-and-Governance constellations, adding shared-state ownership, dissent, checkpoints, and coordination. Day 8 will introduce a meta-agent that plans and delegates across validated registries. Day 9 will integrate the components into a fail-closed autonomous research runtime, and Day 10 will create the governed research-to-LEAN bridge and experimentation framework.

The notebook should therefore be read as an institutional constitution expressed in executable specifications. Its achievement is not that it creates unconstrained intelligence. It creates legible autonomy: every capability has a contract, every procedure has a boundary, every role has an allowlist, every action consumes budget, every material decision has an escalation path, and every tested outcome carries evidence. The durable principle is that autonomy grows through validated composition while authority remains explicit, limited, and human-accountable.


## Workflow: from validated functions to governed institutional roles

The capability factory follows a strict admission sequence. A component is not promoted merely because code exists. It must have a complete specification, valid dependencies, bounded permissions, evidence of testing, an explicit lifecycle state, and a defined place in the handoff architecture.

| 3. Compose skills | Which repeatable procedure combines approved tools safely? | Nine skill specifications | Tool allowlist, evidence requirements, quality checks, boundaries, stopping rules |
| 4. Define agents | Which institutional role may perform which procedures? | 13 bounded-agent specifications | Skill and tool allowlists, memory, budget, permissions, gates, handoffs |

| 7. Hand off to Day 7 | Can roles be coordinated without inventing new permission? | Governed registry package | Validated registries, audit hashes, boundaries, and escalation contracts |

## 4. Verify registry integrity and inherited authority

Registry validation checks that identifiers are unique and that every skill and agent references only registered components. It also verifies that each agent’s tool allowlist is exactly the union of the tools permitted by its assigned skills. Finally, every agent must contain explicit denials for self-approval and live execution. These checks prevent missing dependencies, invented capabilities, privilege drift, and silent expansion of authority.


```python
tool_ids = {x['tool_id'] for x in TOOLS}
skill_ids = {x['skill_id'] for x in SKILLS}
agent_ids = {x['agent_id'] for x in AGENTS}
skill_tools = {x['skill_id']: set(x['allowed_tools']) for x in SKILLS}
dependency_errors = []

if len(tool_ids) != len(TOOLS):
    dependency_errors.append('duplicate_tool_id')
if len(skill_ids) != len(SKILLS):
    dependency_errors.append('duplicate_skill_id')
if len(agent_ids) != len(AGENTS):
    dependency_errors.append('duplicate_agent_id')

for skill in SKILLS:
    missing = set(skill['allowed_tools']) - tool_ids
    if missing:
        dependency_errors.append({'skill': skill['skill_id'], 'missing_tools': sorted(missing)})

for agent in AGENTS:
    missing_skills = set(agent['allowed_skills']) - skill_ids
    missing_tools = set(agent['allowed_tools']) - tool_ids
    inherited_tools = set().union(*(skill_tools[s] for s in agent['allowed_skills']))
    if missing_skills or missing_tools or inherited_tools != set(agent['allowed_tools']):
        dependency_errors.append({
            'agent': agent['agent_id'],
            'missing_skills': sorted(missing_skills),
            'missing_tools': sorted(missing_tools),
            'allowlist_matches_skills': inherited_tools == set(agent['allowed_tools']),
        })

all_components_validated = all(
    x['status'] == 'validated' for collection in (TOOLS, SKILLS, AGENTS) for x in collection
)
all_agents_bounded = all(
    any(p['capability'] == 'approve' and p['effect'] == 'deny' for p in agent['permissions'])
    and any(p['capability'] == 'execute' and p['effect'] == 'deny' for p in agent['permissions'])
    for agent in AGENTS
)

registry_integrity = {
    'dependency_errors': dependency_errors,
    'all_components_validated': all_components_validated,
    'all_agents_deny_self_approval_and_live_execution': all_agents_bounded,
}
assert dependency_errors == []
assert all_components_validated
assert all_agents_bounded
registry_integrity
```

## Frequently Asked Questions — NB03

## 1. What problem does the capability factory solve?

Earlier notebooks contain useful functions, but a function alone does not tell an autonomous system when it may be called, what evidence it requires, which outputs it promises, how it fails, or which effects it is forbidden to produce. The capability factory converts validated functions into governed objects that can be discovered, composed, tested, and audited without treating every piece of code as universally authorized.

## 2. What is the difference between a function and a tool?

A function is executable code. A governed tool is a function surrounded by a specification. In NB03, each tool has an identifier, version, purpose, input and output contracts, errors, permissions, side effects, provenance, tests, and status. The wrapper makes the capability legible to registries, skills, agents, governance controls, and auditors.

## 3. What fields are required in a tool specification?

Each registered tool contains `schema_version`, `tool_id`, `name`, `version`, `purpose`, `inputs`, `outputs`, `errors`, `permissions`, `side_effects`, `provenance`, `tests`, and `status`. Inputs require a request identifier and evidence; outputs require a status and provenance record. The error envelope distinguishes missing evidence from unauthorized action, and the permission list explicitly denies live execution.

## 4. Why are provenance and tests part of the tool definition?

Provenance identifies the validated source from which the capability was promoted and requires a content hash. Tests provide evidence for both the normal path and the denial path. This means registration depends on more than a successful example: the capability must also demonstrate how it fails closed when an unauthorized action is requested.

## 5. What is a skill in this architecture?

A skill is a governed procedure that composes allowed tools toward a defined objective. It adds sequence, evidence requirements, quality checks, boundaries, escalation, and stopping conditions. A skill answers not only “which functions are available?” but also “how should this work be performed responsibly, and when must it stop?”

## 6. Which nine skills are registered?

The notebook registers Data Research, Model Development, Strategy Design, Backtesting, Portfolio Construction, Risk Review, Governance Review, Audit Bundle, and Research to LEAN. Together they cover the research lifecycle while keeping execution, risk, governance, audit, and platform-conversion responsibilities separately visible.

## 7. What makes an agent bounded?

An agent is bounded by its objective, allowed skills, allowed tools, memory scope, step budget, tool-call budget, permissions, policy gates, handoffs, escalation triggers, and lifecycle status. The agent cannot choose arbitrary capabilities merely because they exist elsewhere in the registry. It may act only through the allowlist attached to its institutional role.

## 8. Which agent roles are created?

NB03 registers 13 roles: Research Analyst, Data Scientist, Experiment Designer, Strategy Designer, Portfolio Constructor, Risk Officer, Governance Critic, Audit Reviewer, Execution Simulation Agent, QuantConnect Code Generator, Code Report Generator, Governance Strategy Evaluation Agent, and Research-to-QuantConnect Converter. These are deterministic role specifications, not yet an LLM orchestration runtime.

## 9. How do tool and skill allowlists work?

Each skill names the tools it may invoke. Each agent names its allowed skills and receives the corresponding union of approved tools. The integrity checks confirm that every reference exists and that an agent’s tool set matches its skill assignments. A request for an unregistered tool or a tool outside the role’s allowlist is denied as hallucinated capability or agent overreach.

## 10. What do the memory and budget controls accomplish?

Working memory is limited to the current task envelope, while durable access is read-only. The standard budget allows at most eight steps and 12 tool calls. These limits constrain runaway recursion, uncontrolled accumulation of context, and open-ended resource use. When the budget is exhausted, the correct result is to stop and escalate rather than silently expand the budget.

## 11. Which policy gates must an agent pass?

Every agent is subject to semantic sufficiency, registered-capability validation, evidence completeness, budget availability, separation of duties, and a research-only boundary. Semantic sufficiency asks whether the request contains enough meaning to evaluate. Registered-capability validation prevents invented tools. Evidence completeness requires the artifacts specified by the skill. Budget availability prevents unbounded work. Separation of duties prevents self-approval. The research-only gate denies live activity.

## 12. How are approval and live execution controlled?

All agent specifications explicitly deny approval of their own outputs, promotions, and overrides. They also deny execution in live markets and brokerage systems. A strategy agent may recommend a portfolio, and an execution agent may simulate orders, but neither can authorize promotion or send an order. Material decisions are handed to the human governance owner.

## 13. What is the role of handoffs and escalation?

A handoff transfers a structured evidence package to the role responsible for the next decision. Risk-bearing recommendations go to the Risk Officer. Approval, exception, ambiguity, prohibited requests, missing evidence, exhausted budgets, unregistered capabilities, live-action requests, and self-approval requests escalate to the human governance owner. The payload includes the request, evidence, denial reason, and audit hash.

## 14. How is separation of duties implemented?

Separation of duties is implemented through distinct roles, restricted allowlists, explicit permission denials, policy gates, and mandatory handoffs. The role producing a strategy cannot approve the strategy. The execution role cannot convert simulation into live trading. Risk and governance remain independent challenge functions, and material authorization remains human-accountable.

## 15. How is registry integrity measured?

The notebook verifies unique tool, skill, and agent identifiers; validated lifecycle states; complete skill-to-tool references; complete agent-to-skill and agent-to-tool references; exact inheritance of tool allowlists from assigned skills; and explicit denial of self-approval and live execution. The validated registry contains zero dependency errors.

## 16. What adversarial scenarios are tested?

Seven scenarios are evaluated. A valid research request must complete. Six adverse requests must be denied: hallucinated or unregistered capability, agent overreach, incomplete evidence, budget exhaustion, unauthorized live execution, and self-approval. Every scenario reaches its expected outcome and records an audit hash.

## 17. Why is a denied request considered a successful result?

Success is defined as compliance with the institutional mandate, not universal task completion. If a request lacks evidence, exceeds a role, consumes its budget, violates separation of duties, or asks for live authority, denial protects the system. A blocked proposal is therefore a valid result when blocking is the policy-correct outcome.

## 18. Does `validated` mean production-ready?

No. `Validated` means the component satisfied the notebook’s synthetic-equity, research-only specifications and tests. It does not establish security under hostile production conditions, real-data correctness, broker integration, market capacity, regulatory approval, or live readiness. No claim may exceed the evidence boundary.

## 19. Are these agents autonomous artificial intelligences?

They are bounded, deterministic role definitions that make institutional autonomy inspectable. They do not yet constitute an LLM-based planning and orchestration runtime. Day 7 coordinates roles in constellations, Day 8 adds the governed meta-agent, and Day 9 integrates the system runtime. NB03 supplies the constitutional registry those later layers must obey.

## 20. What is the final lesson of NB03?

Autonomy should grow through validated composition, not through uncontrolled permission. Tools define atomic capability, skills define governed procedure, agents define bounded responsibility, registries make dependencies discoverable, tests verify both action and refusal, and human governance retains material authority. The system becomes more capable while its permissions remain explicit and limited.


## How Agents, Skills, and Tools Are Routed in NB03

## 1. The central routing principle

NB03 does not allow an agent to select any function it finds useful. Every request must travel through a governed route:

$
\text{Request}
\rightarrow
\text{Policy Gates}
\rightarrow
\text{Agent}
\rightarrow
\text{Skill}
\rightarrow
\text{Tool}
\rightarrow
\text{Verified Output}
\rightarrow
\text{Audit Record}
$

The routing architecture separates four concepts:

- **The request** describes what the system is being asked to accomplish.
- **The agent** represents the institutional role responsible for the task.
- **The skill** defines the approved procedure the agent may follow.
- **The tool** performs a specific atomic operation.

An agent cannot invoke a tool merely because the tool exists in the registry. The agent must possess a skill that explicitly exposes that tool.

The effective authority of an action can be expressed as:

$
\text{Effective Authority}
=
\text{System Policy}
\cap
\text{Agent Permissions}
\cap
\text{Skill Allowlist}
\cap
\text{Tool Contract}
\cap
\text{Request Scope}
$

If any layer denies the action, the route stops.

---

## 2. Stage One: the request enters the policy router

Every workflow begins with a structured request. A valid request should identify:

- The requested action
- The target artifact
- The operating scope
- The supporting evidence
- The responsible role
- The required output
- The applicable budget

Before an agent or tool is selected, the request must pass six policy gates.

| Policy gate | Question |
|---|---|
| Semantic sufficiency | Is the request complete and sufficiently unambiguous? |
| Registered capability | Does the requested capability exist in the approved registry? |
| Evidence complete | Are the required data, hashes, assumptions, and artifacts available? |
| Budget available | Are sufficient steps and tool calls available? |
| Separation of duties | Is the requesting role permitted to perform this action? |
| Research only | Does the action remain inside the synthetic research environment? |

A request proceeds only when every applicable gate passes.

If a gate fails, the system must stop, deny, or escalate. It must not reinterpret the request creatively to force completion.

---

## 3. Stage Two: the request is assigned to an agent

The router selects an agent whose institutional objective matches the request.

NB03 defines 13 agent roles:

1. Research Analyst
2. Data Scientist
3. Experiment Designer
4. Strategy Designer
5. Portfolio Constructor
6. Risk Officer
7. Governance Critic
8. Audit Reviewer
9. Execution Simulation Agent
10. QuantConnect Code Generator
11. Code Report Generator
12. Governance Strategy Evaluation Agent
13. Research-to-QuantConnect Converter

Every agent has:

- An institutional objective
- An allowed-skill list
- An allowed-tool list
- A current-task memory boundary
- A maximum of eight steps
- A maximum of 12 tool calls
- Policy gates
- Handoff rules
- Escalation conditions
- Explicit permission denials

All agents are prohibited from approving their own outputs, overriding risk controls, connecting to a broker, or executing live orders.

---

## 4. Stage Three: the agent selects an authorized skill

A skill represents a governed procedure. It tells the agent how to combine tools, what evidence is required, which quality checks must be performed, and when the procedure must stop.

The nine skills are:

| Skill | Principal function |
|---|---|
| Data Research | Prepare and validate point-in-time research data |
| Model Development | Train and diagnose prediction models |
| Strategy Design | Convert predictions into governed investment logic |
| Backtesting | Evaluate strategies using bounded simulation |
| Portfolio Construction | Convert signals into constrained positions |
| Risk Review | Measure exposures, losses, limits, and stresses |
| Governance Review | Validate semantics, permissions, provenance, and auditability |
| Audit Bundle | Assemble reconciliation and governance evidence |
| Research to LEAN | Translate research artifacts toward a LEAN-compatible representation |

The agent may use only the skills included in its specification. A Strategy Designer, for example, cannot simply perform an independent risk review because the `risk_review` skill is not part of its allowlist.

## 5. Stage Four: the skill exposes specific tools

Each skill exposes exactly four registered tools.

For example, the Data Research skill exposes:

- Canonical Data Loader
- Dataset Validator
- Point In Time Feature Builder
- Chronological Splitter

The skill does not grant access to every tool in the registry. Its allowlist creates a bounded operational surface.

Some tools are shared across multiple skills. For example:

- Volatility Targeter is shared by Strategy Design and Portfolio Construction.
- Audit Bundle Builder is shared by Backtesting and Audit Bundle.
- Canonical Data Loader is shared by Data Research and Research to LEAN.
- Signal Generator is shared by Strategy Design and Research to LEAN.
- Provenance Verifier is shared by Governance Review, Audit Bundle, and Research to LEAN.
- Immutable Audit Logger is shared by Governance Review and Audit Bundle.

Shared access does not mean identical responsibility. The same tool may support different procedures under different skill boundaries.

## 6. Registered does not mean routed

NB03 registers 35 tools, but only 29 are currently exposed through skills.

The following six tools are registered but not assigned to a skill:

- Spec Validator
- Synthetic Equity Generator
- Canonical Database Builder
- KNN Baseline Trainer
- Execution Simulator
- Market Impact Estimator

These tools exist in the registry, but agents cannot invoke them through the current skill architecture.

This establishes an important distinction:

> **Registered capability is not the same as authorized capability.**

Before an agent could use one of these tools, the tool would have to be assigned to a validated skill, the skill would have to be assigned to an agent, and the resulting route would have to pass governance and dependency tests.

---

## 7. Stage Five: the tool executes under its contract

Every tool declares:

- Purpose
- Inputs
- Outputs
- Error conditions
- Permissions
- Side effects
- Provenance
- Golden-path test
- Denial-path test
- Lifecycle status

The tool receives only the information permitted by the request, agent, and skill.

A successful tool call should return a structured result containing a status and provenance record. An unsuccessful call should produce an explicit error or denial reason rather than an ambiguous or fabricated output.

The tool cannot expand its own scope. A research tool cannot convert itself into a live execution tool merely because the requested calculation succeeded.

---

## 8. Stage Six: outputs are verified and audited

After a tool produces an output, the system verifies:

- Output-schema compliance
- Evidence completeness
- Provenance
- Content hashes
- Permission compliance
- Budget consumption
- Remaining unresolved questions
- Whether a handoff is required

A hash-bearing audit record documents the request, agent, skill, tool, result, and policy decision.

Risk-bearing recommendations are handed to the Risk Officer. Approval requests, exceptions, ambiguity, prohibited actions, and unresolved governance matters are escalated to the Human Governance Owner.

# Final Routing Doctrine

The NB03 routing architecture can be summarized as:

> **Requests do not go directly to tools. Requests pass through policy gates, institutional roles, governed skills, explicit tool contracts, output verification, and audit.**

The system follows five non-negotiable principles:

1. **A tool must be registered before it can be considered.**
2. **A registered tool must be exposed through a validated skill before an agent can use it.**
3. **The skill must appear in the agent’s allowlist.**
4. **The request must remain within the agent’s permissions, evidence, scope, and budget.**
5. **Approval, exceptions, self-promotion, broker access, and live execution remain outside agent authority.**

The result is an architecture in which autonomy grows through controlled composition:

$
\text{Validated Function}
\rightarrow
\text{Governed Tool}
\rightarrow
\text{Repeatable Skill}
\rightarrow
\text{Bounded Agent}
\rightarrow
\text{Controlled Handoff}
$

> **Agents may recommend and request. Risk challenges. Governance controls. Audit records. Humans remain accountable for material decisions.**
