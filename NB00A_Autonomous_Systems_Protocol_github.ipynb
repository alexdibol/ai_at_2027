## Introduction: writing the constitution before building the machine

NB00A begins with a premise that is easy to state and difficult to honor: an autonomous financial system must be governed before it becomes intelligent. Most technical courses begin with data, predictions, or trading rules and discuss controls later. This course reverses that order. Before a model is permitted to classify a market regime, recommend a position, call a backtest, or communicate with another agent, the platform must define what each component is, which evidence it must provide, which actions it may perform, what errors it can return, when it must stop, and when a human decision is compulsory. This notebook therefore creates the constitutional layer of the entire course. It does not predict prices. It establishes the grammar through which every later capability can be described, tested, registered, challenged, and constrained.

The distinction between capability and authority is the notebook’s central idea. A function may be capable of computing a portfolio weight without possessing authority to allocate real capital. A model may be capable of generating a probability without having authority to convert it into an order. An agent may be capable of planning a sequence of tool calls without being permitted to access credentials, alter durable records, or cross a risk limit. Treating those distinctions as informal intentions is inadequate because complex systems eventually encounter ambiguous requests, missing data, contradictory objectives, and novel states. NB00A translates intentions into machine-readable contracts so that permissions and boundaries travel with the component rather than depending on memory or goodwill.

The course builds upward through a deliberate sequence: environment → data → models → tools → skills → agents → agent constellations → meta-agent → autonomous system. The five-act narrative expresses the same architecture institutionally. Act I, **Build the World**, establishes protocols, data, provenance, and reproducibility. Act II, **Build Intelligence**, develops models, strategies, and portfolio policies. Act III, **Build Knowledge**, turns validated functions into reusable tools and combines tools into governed skills. Act IV, **Build the Organization**, assigns those skills to bounded research, risk, audit, and coordination agents. Act V, **Build the Machine**, permits agents to cooperate under a meta-agent while preserving traceability and human accountability. NB00A is the first notebook because every downstream layer inherits its definitions.

The notebook operationalizes that architecture through eight canonical schemas. `ToolSpec` defines a deterministic capability through its purpose, inputs, outputs, errors, permissions, side effects, provenance, tests, and lifecycle status. `SkillSpec` describes a repeatable procedure that may coordinate several tools while stating evidence requirements, quality checks, boundaries, escalation, and stopping conditions. `AgentSpec` adds a role, objective, memory scope, budgets, permissions, policy gates, and handoff rules. `SharedState` creates a controlled record through which multiple agents can exchange observations, assumptions, decisions, artifacts, questions, approvals, and audit references. These contracts prevent a system from calling every piece of code a tool or every prompt an agent.

The remaining schemas make accountability durable. `ProvenanceEnvelope` ties an artifact to its producer, source identifiers, code version, environment, parameters, and cryptographic content hash. `HumanApproval` records the scope, evidence, state, conditions, and responsible decision-maker for an approval. `AuditEvent` captures the ordered sequence of observations, tool calls, decisions, handoffs, approvals, exceptions, and terminations. `RegistryEntry` identifies the validated artifact, its version, status, implementation, test report, dependencies, and hash. Together, these objects form an evidence chain. A result is not trustworthy merely because it exists; it becomes inspectable when the system can explain where it came from, under which rules it was produced, and whether those rules were satisfied.

Schemas alone do not create governance. A constitution that is never enforced is only prose. NB00A therefore builds a pure local validator that checks candidate documents against the canonical contracts and returns standardized, path-specific errors. The validator is deliberately narrow: it reads schemas and documents, performs structural checks, and may write a validation report. It has no network, shell, credential, brokerage, or trading permission. This design anticipates the tool architecture used throughout the course. Deterministic components should do bounded work predictably; interpretive components may explain or challenge results but may not silently expand the deterministic component’s authority.

The testing logic is equally important. Three valid examples—a tool, a skill, and an agent—must pass. Three defective examples must fail for the intended reasons: a tool without permissions, a skill without boundaries, and an agent with a zero-step budget. Positive tests prove that acceptable objects can pass, while negative tests prove that invalid objects are actually rejected. This is fail-closed governance. If a schema is unknown, a required field is missing, a boundary is empty, or a budget violates its minimum, the object is not registered by default. Uncertainty does not become permission merely because a workflow wishes to continue.

The policy-gate section moves from structural validity to action admissibility. Semantic sufficiency asks whether a request contains enough information to evaluate responsibly. The action rule then returns one of three outcomes: allow a bounded non-material action, deny an action that exceeds the system’s authority, or require human approval for a material action. This three-way logic is more useful than a simple yes/no switch because financial work contains decisions that are legitimate only when an accountable person has reviewed specified evidence. Human approval is not a ceremonial final click; it is a governed state with scope, conditions, and revocability.

Finally, the notebook exports the validator as a reusable tool, writes its `ToolSpec`, records validation evidence, calculates a content hash, and creates a registry entry. That closing sequence establishes the pattern that every later notebook must follow: educational exposition leads to deterministic implementation; implementation is challenged by tests; validated behavior is documented through a machine-readable contract; and the resulting capability is registered without receiving powers it does not need. The model proposes, governance challenges, audit evaluates, and the human remains accountable. NB00A is therefore not administrative overhead before the “real” quantitative work. It is the mechanism that makes advanced quantitative and agentic work capable of entering a professional financial institution.

## 3. Build the validation tool

The validator recursively checks types, constants, enumerations, minimum lengths, numeric floors, uniqueness, required fields, and additional properties. Its response is deliberately standardized: a validity flag, the selected schema, and actionable errors containing a field path, message, and code. It remains pure and local, with no network, shell, credential, brokerage, or live-trading authority.


```python
from __future__ import annotations
import json
import re
from pathlib import Path
from typing import Any

def _type_ok(value, expected):
    if isinstance(expected, list): return any(_type_ok(value, e) for e in expected)
    return {"object": isinstance(value, dict), "array": isinstance(value, list),
            "string": isinstance(value, str), "integer": isinstance(value, int) and not isinstance(value, bool),
            "boolean": isinstance(value, bool), "null": value is None}.get(expected, True)

def _walk(schema, value, path="$"):
    errors = []
    if "type" in schema and not _type_ok(value, schema["type"]):
        return [{"path": path, "message": f"Expected {schema['type']}", "code": "VALIDATION_ERROR"}]
    if "const" in schema and value != schema["const"]: errors.append({"path": path, "message": f"Must equal {schema['const']}", "code": "VALIDATION_ERROR"})
    if "enum" in schema and value not in schema["enum"]: errors.append({"path": path, "message": f"Must be one of {schema['enum']}", "code": "VALIDATION_ERROR"})
    if isinstance(value, str):
        if len(value) < schema.get("minLength", 0): errors.append({"path": path, "message": "String is too short", "code": "VALIDATION_ERROR"})
        if "pattern" in schema and not re.match(schema["pattern"], value): errors.append({"path": path, "message": "String does not match pattern", "code": "VALIDATION_ERROR"})
    if isinstance(value, (int, float)) and not isinstance(value, bool) and value < schema.get("minimum", value): errors.append({"path": path, "message": f"Must be >= {schema['minimum']}", "code": "VALIDATION_ERROR"})
    if isinstance(value, list):
        if len(value) < schema.get("minItems", 0): errors.append({"path": path, "message": "Array has too few items", "code": "VALIDATION_ERROR"})
        if schema.get("uniqueItems") and len({json.dumps(x, sort_keys=True) for x in value}) != len(value): errors.append({"path": path, "message": "Array items must be unique", "code": "VALIDATION_ERROR"})
        for i, item in enumerate(value): errors.extend(_walk(schema.get("items", {}), item, f"{path}[{i}]"))
    if isinstance(value, dict):
        for key in schema.get("required", []):
            if key not in value: errors.append({"path": f"{path}.{key}", "message": "Required property is missing", "code": "VALIDATION_ERROR"})
        props = schema.get("properties", {})
        if schema.get("additionalProperties") is False:
            for key in value:
                if key not in props: errors.append({"path": f"{path}.{key}", "message": "Additional property is not allowed", "code": "VALIDATION_ERROR"})
        for key, item in value.items():
            if key in props: errors.extend(_walk(props[key], item, f"{path}.{key}"))
    return errors

class SpecValidator:
    """Pure, local validator. It has no network, shell, or trading authority."""
    def __init__(self, schema_dir: str | Path):
        self.schema_dir = Path(schema_dir)
        self.schemas = {p.name.replace(".schema.json", ""): json.loads(p.read_text(encoding="utf-8")) for p in self.schema_dir.glob("*.schema.json")}

    def validate(self, schema_name: str, document: dict[str, Any]) -> dict[str, Any]:
        if schema_name not in self.schemas:
            return {"valid": False, "schema": schema_name, "errors": [{"path": "$", "message": "Unknown schema", "code": "UNKNOWN_SCHEMA"}]}
        errors = _walk(self.schemas[schema_name], document)
        return {"valid": not errors, "schema": schema_name, "errors": errors}

    def validate_file(self, schema_name: str, document_path: str | Path) -> dict[str, Any]:
        document = json.loads(Path(document_path).read_text(encoding="utf-8"))
        return self.validate(schema_name, document)

validator = SpecValidator(ROOT/'schemas')
print('Validator ready with', len(validator.schemas), 'schemas')
```

## 5. Exercise semantic and authority policy gates

Structural validity does not imply authorization. The policy layer first checks whether a request is sufficiently specified, then classifies the proposed action as allowed, denied, or subject to human approval. A bounded documentation action may proceed; live financial execution is denied; a material research publication requires an accountable decision. The returned reason is part of the evidence, not an informal comment.


```python
def governance_gate(request):
    required = {'objective','evidence_ids','requested_capability','scope'}
    missing = sorted(required - set(request))
    if missing: return {'decision':'deny','reason':'insufficient_semantics','missing':missing}
    if request['requested_capability'] == 'execute': return {'decision':'deny','reason':'research_system_has_no_live_execution_authority'}
    if request['requested_capability'] in {'write','approve'}: return {'decision':'require_approval','reason':'material_action'}
    return {'decision':'allow','reason':'bounded_non_material_action'}

for r in [
 {'objective':'inspect schema','evidence_ids':['schema_v1'],'requested_capability':'read','scope':'local'},
 {'objective':'place order','evidence_ids':['signal_1'],'requested_capability':'execute','scope':'broker'},
 {'objective':'change registry','evidence_ids':['test_1'],'requested_capability':'write','scope':'validated_registry'}]:
    print(governance_gate(r))
```
