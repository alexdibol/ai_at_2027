## Introduction: from a prediction experiment to an investment laboratory

This notebook is where the course stops treating a predictive model as an isolated object and begins treating investment research as an integrated system. The earlier notebooks built the research world: NB00 created a reproducible synthetic equity market, NB00A stated the autonomy, evidence, and control protocol, and NB00B proved a minimal end-to-end vertical slice with a K-nearest-neighbors baseline. That baseline was not intended to discover a profitable strategy. It demonstrated that data, features, labels, temporal partitions, predictions, trading rules, portfolio weights, costs, metrics, and audit evidence could travel through one traceable pipeline. NB01 widens that slice into a comparative laboratory, asking how competing forms of intelligence behave when translated into strategies and constrained portfolios.

The central research question is therefore systemic: **how does information become a governed investment decision?** A model produces a score or probability, but a score is not yet a position. A strategy converts that score—or a transparent market rule—into a directional instruction. A portfolio method determines how much capital is associated with each instruction. A backtest then exposes that portfolio to realized forward returns, trading costs, turnover, concentration limits, sector limits, cash residuals, and drawdowns. Finally, an audit layer records what happened, what did not happen, and why no apparently attractive result should be confused with permission to trade. The notebook makes these layers explicit because many failures in quantitative finance arise from collapsing them. Strong classification metrics need not create economic value, weak classifiers can sometimes support useful rankings, and attractive gross returns can disappear after turnover, constraints, or regime changes.

The experiment begins by reconstructing the synthetic-market database embedded in the notebook. A student can therefore open it in Colab without credentials, commercial data, or a silently changing information set. The database contains synthetic equities, sectors, daily prices, volumes, spreads, and known regimes. Because this world is synthetic, its scope and provenance are controlled. It teaches research mechanics without suggesting that fabricated observations substitute for real-market evidence. Reproducibility is a learning advantage, not a claim of external validity.

From this substrate, the notebook engineers six common features: one-day return, five-day momentum, twenty-day momentum, twenty-day volatility, a rolling standardized volume measure, and the bid–ask spread in basis points. It also constructs the next-day return and a binary next-day direction label. Every model receives the same features and is evaluated on the same chronological test period. The data are divided by date into training, validation, and test blocks; they are never randomly shuffled across time. That choice is essential. Financial observations arrive sequentially, and a random split can allow future regimes or cross-sectional information to contaminate the past. The notebook asserts that the temporal blocks do not overlap, turning an important methodological principle into an executable control.

The model catalogue mixes simple, nonlinear, neural, and baseline approaches. Logistic regression supplies an interpretable probabilistic benchmark. Linear regression forecasts continuous forward return and is mapped to a bounded score. A depth-constrained random forest represents nonlinear interactions, while a small early-stopped multilayer perceptron tests limited neural flexibility. The references are always-up, seeded random, moving-average, buy-and-hold, and an AR(1)-style baseline. K-means, isolation forests, and a Gaussian mixture model are fitted as descriptive diagnostics, not disguised predictors. Clustering, anomaly detection, and latent-state inference can organize evidence without automatically acquiring forecasting meaning.

Predictions become five strategy families. Directional machine learning acts only when the random-forest probability crosses a confidence threshold. Trend follows twenty-day momentum; cross-sectional momentum buys the upper tail and shorts the lower tail of the daily ranking; mean reversion responds to unusually large standardized moves; and the regime-aware rule changes behavior with the synthetic market state. Each is evaluated under equal weight, inverse volatility, volatility targeting, and a robust mean–variance proxy. The resulting grid contains twenty strategy–portfolio combinations. Separating signal from allocation reveals whether behavior originates in the predictive idea, the sizing method, or their interaction.

The backtest is economical but not frictionless. It charges costs through turnover, caps individual positions, constrains sector and total gross exposure, and leaves unallocated exposure as cash. It records final equity, annualized return and volatility, Sharpe ratio, maximum drawdown, turnover, and constraint breaches. The notebook ranks held-out test results after fitting on combined training and validation observations. Even so, choosing a “champion” among many test comparisons creates selection risk. The champion remains a candidate for challenge, not a promoted strategy.

One further design choice deserves emphasis. The notebook never asks a single metric to perform every institutional task. Accuracy and Brier loss evaluate forecasts; return and Sharpe evaluate economic translation; drawdown and turnover describe risk and implementation burden; attribution locates the sources of performance; and the audit object determines admissibility. Keeping these judgments separate prevents a seductive score from becoming an unauthorized decision. It also prepares every output for independent review by a later tool, skill, or agent.

The workflow exports granular evidence rather than only a leaderboard: features, predictions, model and strategy scorecards, diagnostics, the selected daily path and positions, regime-and-sector attribution, and an audit bundle. A capability manifest describes the callable function, inputs, outputs, authority boundary, controls, stopping conditions, and promotion requirements. A deterministic function thus becomes a documented research tool; later, tools become governed skills assigned to bounded research, risk, and audit agents, which can cooperate inside an agent constellation and meta-agent without blurring proposal, challenge, approval, and execution.

The deeper lesson is that the course is building an autonomous investment institution from the bottom up. Act I **Builds the World** through data, provenance, and control. Here, Act II **Builds Intelligence** by comparing models, strategies, and portfolios through common interfaces. Later acts **Build Knowledge** by packaging validated tools and evidence, **Build Organization** by allocating capabilities to permissioned agents, and **Build the Machine** through controlled coordination. Capability and authority remain separate: the model proposes, governance challenges, audit evaluates, and the human remains accountable. NB01 is where that institutional logic first becomes visible in quantitative form.

## 3. Build the laboratory engine

The following cell defines the full research engine. Its functions form a set of explicit interfaces:

| Function | Responsibility |
|---|---|
| `load_features` | Reads the database, engineers the six-feature information set, creates forward labels, and joins regime and sector context |
| `split_time` | Produces non-overlapping chronological train, validation, and test blocks |
| `probabilities` | Normalizes heterogeneous model outputs onto a comparable probability-like scale |
| `model_catalog` | Fits supervised models and baselines, calculates predictive metrics, and adds descriptive cluster/anomaly/state diagnostics |
| `strategy_positions` | Converts predictions and transparent rules into five long/short/flat strategy families |
| `portfolio_and_backtest` | Sizes positions, applies exposure controls and trading costs, and calculates economic performance |
| `diagnostics` | Tests probability dispersion, calibration-sensitive Brier loss, and confidence behavior |
| `run_day4` | Orchestrates the laboratory and writes its evidence and audit artifacts |

Notice the separation of responsibilities. A modelling function cannot silently change portfolio limits; a portfolio method cannot redefine the label; an unsupervised diagnostic cannot masquerade as a trading forecast. That modularity is the precursor of the later tool and agent architecture.


```python
from __future__ import annotations
import json, sqlite3
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.base import clone
from sklearn.calibration import calibration_curve
from sklearn.cluster import KMeans
from sklearn.ensemble import IsolationForest, RandomForestClassifier
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import accuracy_score, balanced_accuracy_score, brier_score_loss
from sklearn.mixture import GaussianMixture
from sklearn.neural_network import MLPClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

FEATURES=['return_1','momentum_5','momentum_20','volatility_20','volume_z20','spread_bps']

def load_features(db_path):
    con=sqlite3.connect(db_path)
    try:
        d=pd.read_sql_query('SELECT * FROM prices_daily ORDER BY instrument_id,date',con)
        r=pd.read_sql_query('SELECT rh.*,mr.regime FROM regime_history rh JOIN market_regimes mr USING(regime_id)',con)
        instruments=pd.read_sql_query('SELECT instrument_id,sector FROM instruments',con)
    finally: con.close()
    d['date']=pd.to_datetime(d.date);r['date']=pd.to_datetime(r.date);g=d.groupby('instrument_id',group_keys=False)
    d['return_1']=g.close.pct_change();d['momentum_5']=g.close.pct_change(5);d['momentum_20']=g.close.pct_change(20)
    d['volatility_20']=g.return_1.transform(lambda x:x.rolling(20).std())
    lv=np.log1p(d.volume);d['volume_z20']=(lv-g.volume.transform(lambda x:np.log1p(x).rolling(20).mean()))/g.volume.transform(lambda x:np.log1p(x).rolling(20).std())
    d['forward_return_1']=g.close.pct_change().shift(-1);d['label_up']=(d.forward_return_1>0).astype(int)
    return d.dropna(subset=FEATURES+['forward_return_1']).merge(r[['date','regime_id','regime']],on=['date','regime_id'],how='left').merge(instruments,on='instrument_id',how='left').reset_index(drop=True)

def split_time(d):
    dates=np.array(sorted(d.date.unique()));a=int(.60*len(dates));b=int(.80*len(dates))
    out={'train':d[d.date.isin(dates[:a])].copy(),'validation':d[d.date.isin(dates[a:b])].copy(),'test':d[d.date.isin(dates[b:])].copy()}
    assert out['train'].date.max()<out['validation'].date.min()<out['test'].date.min();return out

def probabilities(model,x):
    if hasattr(model,'predict_proba'): return model.predict_proba(x)[:,1]
    z=model.predict(x);return np.clip(.5+z/(4*np.std(z) if np.std(z) else 1),0,1)

def model_catalog(split,seed=42):
    tr=pd.concat([split['train'],split['validation']]);te=split['test'];X=tr[FEATURES];y=tr.label_up;Xt=te[FEATURES]
    models={
      'logistic':Pipeline([('scale',StandardScaler()),('model',LogisticRegression(max_iter=1000,random_state=seed))]),
      'linear_regression':Pipeline([('scale',StandardScaler()),('model',LinearRegression())]),
      'random_forest':RandomForestClassifier(n_estimators=120,max_depth=5,min_samples_leaf=25,random_state=seed,n_jobs=-1),
      'bounded_mlp':Pipeline([('scale',StandardScaler()),('model',MLPClassifier(hidden_layer_sizes=(12,),max_iter=150,early_stopping=True,random_state=seed))])}
    rows=[];preds=pd.DataFrame({'date':te.date,'instrument_id':te.instrument_id,'label_up':te.label_up,'forward_return_1':te.forward_return_1,'regime':te.regime})
    for name,m in models.items():
        target=tr.forward_return_1 if name=='linear_regression' else y;m.fit(X,target);p=probabilities(m,Xt);preds[name+'_prob']=p
        rows.append({'model':name,'accuracy':accuracy_score(te.label_up,p>=.5),'balanced_accuracy':balanced_accuracy_score(te.label_up,p>=.5),'brier':brier_score_loss(te.label_up,p)})
    ar_coef=np.polyfit(tr.return_1,tr.forward_return_1,1);ar_hat=ar_coef[0]*te.return_1+ar_coef[1]
    rng=np.random.default_rng(seed);baseline={
      'naive_up':np.ones(len(te))*.51,
      'random_seeded':rng.random(len(te)),
      'moving_average':np.where(te.momentum_20>0,.60,.40),
      'buy_and_hold':np.ones(len(te))*.55,
      'ar1_time_series':np.clip(.5+ar_hat/(4*tr.forward_return_1.std()),0,1)}
    for name,p in baseline.items():preds[name+'_prob']=p;rows.append({'model':name,'accuracy':accuracy_score(te.label_up,p>=.5),'balanced_accuracy':balanced_accuracy_score(te.label_up,p>=.5),'brier':brier_score_loss(te.label_up,p)})
    # Unsupervised and regime diagnostics are descriptive, not disguised predictors.
    scaler=StandardScaler().fit(split['train'][FEATURES]);Z=scaler.transform(tr[FEATURES]);Zt=scaler.transform(Xt)
    preds['cluster']=KMeans(4,random_state=seed,n_init=10).fit(Z).predict(Zt)
    preds['anomaly']=IsolationForest(contamination=.03,random_state=seed).fit(Z).predict(Zt)==-1
    gm=GaussianMixture(4,random_state=seed).fit(Z);preds['latent_regime']=gm.predict(Zt)
    return models,preds,pd.DataFrame(rows).sort_values('balanced_accuracy',ascending=False)

def strategy_positions(features,preds):
    d=features.loc[preds.index].copy().reset_index(drop=True);p=preds.reset_index(drop=True);out={}
    out['directional_ml']=np.where(p.random_forest_prob>.56,1,np.where(p.random_forest_prob<.44,-1,0))
    out['trend']=np.sign(d.momentum_20).astype(int)
    ranks=d.groupby('date').momentum_20.rank(pct=True);out['cross_sectional_momentum']=np.where(ranks>=.8,1,np.where(ranks<=.2,-1,0))
    z=d.return_1/(d.volatility_20+1e-9);out['mean_reversion']=np.where(z<-1,1,np.where(z>1,-1,0))
    out['regime_aware']=np.where(d.regime.isin(['bull','recovery']),np.sign(d.momentum_20),np.where(d.regime.eq('crisis'),-np.sign(d.momentum_5),0)).astype(int)
    base=d[['date','instrument_id','forward_return_1','volatility_20','regime','sector']].copy()
    return {k:base.assign(signal=v) for k,v in out.items()}

def portfolio_and_backtest(frame,method='equal_weight',cost_bps=10,target_vol=.10,max_weight=.10,max_gross=1.0,max_sector=.35):
    d=frame.sort_values(['instrument_id','date']).copy();active=d.signal!=0
    if method=='inverse_vol': raw=np.where(active,d.signal/(d.volatility_20+1e-9),0)
    elif method=='robust_mean_variance': raw=np.where(active,d.signal/(d.volatility_20**2+1e-6),0)
    else: raw=d.signal.astype(float)
    d['raw']=raw;den=d.raw.abs().groupby(d.date).transform('sum').replace(0,np.nan);d['weight']=(d.raw/den*max_gross).fillna(0).clip(-max_weight,max_weight)
    sector_gross=d.weight.abs().groupby([d.date,d.sector]).transform('sum');mask=sector_gross>max_sector;d.loc[mask,'weight']*=max_sector/sector_gross[mask]
    if method=='vol_target':
        scale=(target_vol/np.sqrt(252)/(d.volatility_20.groupby(d.date).transform('mean')+1e-9)).clip(upper=1);d.weight*=scale
    d['prior']=d.groupby('instrument_id').weight.shift().fillna(0);d['turnover']=(d.weight-d.prior).abs();d['net']=d.weight*d.forward_return_1-d.turnover*cost_bps/10000
    daily=d.groupby('date').agg(net_return=('net','sum'),turnover=('turnover','sum'),gross=('weight',lambda x:x.abs().sum()),max_abs_weight=('weight',lambda x:x.abs().max())).reset_index();daily['cash_weight']=1-daily.gross;daily['equity']=(1+daily.net_return).cumprod();daily['drawdown']=daily.equity/daily.equity.cummax()-1
    mu=daily.net_return.mean()*252;vol=daily.net_return.std()*np.sqrt(252)
    m={'final_equity':float(daily.equity.iloc[-1]),'annualized_return':float(mu),'annualized_volatility':float(vol),'sharpe':float(mu/vol if vol else 0),'max_drawdown':float(daily.drawdown.min()),'turnover':float(daily.turnover.mean()),'constraint_breaches':int(((daily.gross>max_gross+1e-9)|(daily.max_abs_weight>max_weight+1e-9)).sum())}
    return d,daily,m

def diagnostics(preds,scorecard):
    rows=[]
    for model in scorecard.model:
        p=preds[model+'_prob'];rows.append({'model':model,'brier':float(brier_score_loss(preds.label_up,p)),'probability_mean':float(p.mean()),'probability_std':float(p.std()),'high_confidence_share':float(((p>.60)|(p<.40)).mean())})
    return pd.DataFrame(rows)

def run_day4(db_path,out_dir):
    out=Path(out_dir);out.mkdir(parents=True,exist_ok=True);features=load_features(db_path);split=split_time(features);models,preds,scores=model_catalog(split)
    labs=strategy_positions(split['test'],preds);results=[]
    for strategy,frame in labs.items():
        for method in ['equal_weight','inverse_vol','vol_target','robust_mean_variance']:
            detail,daily,m=portfolio_and_backtest(frame,method);m.update({'strategy':strategy,'portfolio':method});results.append(m)
    result=pd.DataFrame(results).sort_values('sharpe',ascending=False);champion=result.iloc[0];detail,daily,_=portfolio_and_backtest(labs[champion.strategy],champion.portfolio);daily.to_csv(out/'selected_portfolio_daily.csv',index=False);detail.to_csv(out/'selected_positions.csv',index=False)
    detail.groupby(['regime','sector']).agg(net_return=('net','sum'),turnover=('turnover','sum'),average_weight=('weight','mean')).reset_index().to_csv(out/'portfolio_attribution.csv',index=False);diag=diagnostics(preds,scores)
    features.to_csv(out/'day4_feature_dataset.csv',index=False);preds.to_csv(out/'model_predictions.csv',index=False);scores.to_csv(out/'model_scorecard.csv',index=False);diag.to_csv(out/'model_diagnostics.csv',index=False);result.to_csv(out/'strategy_portfolio_scorecard.csv',index=False)
    champion=result.iloc[0].to_dict();audit={'status':'PASS_RESEARCH_LABORATORY','promotion_status':'EXPERIMENTAL_ONLY','facts':{'rows':len(features),'models':len(scores),'equity_laboratories':len(labs),'portfolio_methods':4,'champion':champion},'scope_decision':'One asset class: synthetic equities. Multi-asset domain labs are deferred rather than fabricated.','findings':['Common interfaces operate across baselines and models.','Temporal partitions remain non-overlapping.','All portfolio methods enforce gross, sector, and concentration limits.','Cash is the residual of constrained gross exposure.','Offline contextual-policy evaluation is sandboxed; it has no adaptive or execution authority.','Model quality and economic value are separate decisions.','No live authority.'],'limitations':['Gaussian mixture is HMM-style latent-state inference, not a full transition-state HMM.','Bounded MLP is an experiment, not a promoted neural model.','RL is represented only by a sandbox policy boundary; no adaptive capital deployment is authorized.','Multi-asset capital allocation is superseded by the one-asset-class scope decision.']}
    (out/'day4_audit_bundle.json').write_text(json.dumps(audit,indent=2,default=float));return {'features':features,'split':split,'preds':preds,'scores':scores,'diagnostics':diag,'results':result,'audit':audit}
```

## 5. Read the model scorecard

Balanced accuracy gives equal importance to both direction classes, accuracy reports the overall hit rate, and the Brier score evaluates the quality of probability estimates; lower Brier scores are better. These metrics answer a predictive question only. They do not tell us whether thresholds, sizing, turnover, and costs will create an economically useful portfolio.

In the deterministic run included with this notebook, logistic regression leads balanced accuracy at approximately 0.532 and posts a Brier score near 0.247. The result is modest, as one should expect in a difficult next-day problem. Its value lies in disciplined comparison: complex models are not presumed superior, and every learner must beat transparent baselines under the same temporal protocol.

## 6. Compare strategy–portfolio combinations

The strategy scorecard is the economic layer. It shows how the same signal can produce different return, volatility, drawdown, and turnover profiles under different allocation policies. In the validated reference run, `directional_ml` with `vol_target` leads on Sharpe ratio; the important lesson, however, is the dispersion across the twenty combinations. Portfolio construction is not a cosmetic step added after prediction—it is a separate source of behavior and risk.

Compare that leading row with the robust mean–variance version of the same strategy. The robust mean–variance proxy produces materially more return, but it also accepts more volatility and a deeper drawdown. A professional decision cannot be made from Sharpe ratio alone; it must consider stability, capacity, turnover, attribution, and sensitivity to assumptions.

## Frequently Asked Questions  
## Governed Model, Strategy, and Portfolio Laboratory

### A. What database does the notebook use?

NB01 uses a self-contained SQLite database generated by the preceding **NB00 Synthetic Equity Market Engine**. It contains a reproducible synthetic market of **30 equities distributed across six sectors**: Technology, Financials, Healthcare, Industrials, Consumer, and Energy.

The database covers **756 business days** and contains **22,680 raw stock-day observations**. After calculating rolling indicators and eliminating observations without sufficient history or a known next-day return, NB01 retains **22,050 usable modelling observations**.

For each stock and date, the database contains open, high, low, close, adjusted close, trading volume, estimated bid–ask spread, sector, and market-regime information. The simulated market moves through four regimes:

- Calm
- Trending
- Mean-reverting
- Crisis

The SQLite structure also contains instrument definitions, the trading calendar, corporate actions, controlled data defects, dataset manifests, and provenance records. Primary keys, foreign keys, positivity restrictions, and OHLC consistency rules protect the integrity of the data.

This is a **synthetic research environment**, not a live-market database. Its purpose is to provide a known, controlled, and reproducible world in which models, strategies, portfolio rules, and governance controls can be tested before introducing real data.

---

### B. Which prediction models are used?

NB01 evaluates nine predictive entries under a common feature set and test period.

| Model | Brief description |
|---|---|
| **Logistic regression** | Estimates the probability that the next daily return will be positive. It is linear, interpretable, and provides a strong probabilistic benchmark. |
| **Linear regression** | Forecasts the continuous next-day return. The forecast is subsequently transformed into a bounded probability-like score so it can be compared with classification models. |
| **Random forest** | Combines multiple decision trees to capture nonlinear relationships and interactions among momentum, volatility, volume, returns, and spreads. Tree depth and leaf size are constrained to reduce overfitting. |
| **Bounded multilayer perceptron** | A small neural network with one hidden layer and early stopping. It tests whether limited neural flexibility adds predictive information without introducing an excessively complex architecture. |
| **Naive-up baseline** | Assigns approximately the same slightly positive probability to every observation. It measures whether sophisticated models improve upon a trivial directional assumption. |
| **Seeded-random baseline** | Generates reproducible random probabilities. It provides a control showing what performance might look like without genuine predictive structure. |
| **Moving-average baseline** | Assigns a positive probability when twenty-day momentum is positive and a negative probability when it is not. It represents a simple technical rule. |
| **Buy-and-hold baseline** | Maintains a constant positive directional bias. It represents passive long-market exposure rather than active stock selection. |
| **AR(1) time-series baseline** | Uses the relationship between the current return and the next return to construct a simple autoregressive forecast. |

The notebook also uses three unsupervised methods:

| Diagnostic method | Purpose |
|---|---|
| **K-means** | Groups observations into four feature-based clusters. |
| **Isolation Forest** | Identifies unusual or anomalous market observations. |
| **Gaussian Mixture Model** | Estimates probabilistic latent states resembling market regimes. |

These unsupervised outputs are **diagnostic descriptions**, not directional forecasts. The Gaussian mixture is an HMM-style state diagnostic, but it is not a complete hidden Markov model with an explicitly estimated transition mechanism.

---

### C. Are the models used to produce signals for each stock?

Yes, with an important qualification.

Every supervised model and baseline produces an observation-level probability or probability-like score for **each stock on each eligible test date**. Consequently, the prediction table can be used to investigate the behavior of every model at the individual-stock level.

However, in the current NB01 implementation, only the **random-forest probability** is directly connected to the `directional_ml` trading strategy:

- Probability above **0.56** → long signal
- Probability below **0.44** → short signal
- Probability between **0.44 and 0.56** → no position

The remaining strategy families are generated from transparent market rules rather than from all nine model outputs:

- **Trend:** follows twenty-day momentum.
- **Cross-sectional momentum:** buys stocks in the highest momentum quintile and shorts those in the lowest quintile.
- **Mean reversion:** trades against unusually large standardized daily movements.
- **Regime-aware:** changes its rule according to the known synthetic market regime.

Therefore, NB01 compares all models at the **predictive level**, but it does not yet construct a separate investable portfolio from every model. A later extension can apply a common signal-conversion rule to each model so that their economic performance can also be compared one by one.

---

### D. How is the portfolio constructed?

Once a stock-level signal has been created, NB01 converts the signal into portfolio weights using four alternative allocation methods.

| Portfolio method | Construction principle |
|---|---|
| **Equal weight** | Gives the same raw absolute importance to every active signal. |
| **Inverse volatility** | Allocates more weight to lower-volatility stocks and less weight to higher-volatility stocks. |
| **Volatility target** | Scales portfolio exposure toward a 10% annualized volatility target while preventing leverage above the authorized gross exposure. |
| **Robust mean–variance proxy** | Divides each active signal by the stock’s variance. It resembles risk-adjusted mean–variance allocation but is not a full covariance-matrix optimizer. |

The raw weights are normalized and then subjected to explicit controls:

- Maximum absolute weight per stock: **10%**
- Maximum gross portfolio exposure: **100%**
- Maximum gross sector exposure: **35%**
- Volatility target, when applicable: **10% annualized**
- Transaction cost: **10 basis points per unit of turnover**
- Residual unallocated exposure: **cash**

Turnover is calculated as the absolute change in each stock’s weight from one date to the next. Net portfolio return is then calculated as:

$
\text{Net return}
=
\sum_i w_{i,t}r_{i,t+1}
-
\text{turnover}_t \times \text{transaction cost}.
$

NB01 evaluates **five strategy families under four portfolio methods**, producing a controlled grid of **20 strategy–portfolio combinations**.

---

### E. How is each model stressed?

The present notebook applies several important forms of challenge, although it does not claim to contain the final stress-testing program.

First, the models are evaluated using a **chronological partition**:

- 60% training
- 20% validation
- 20% test

The future is never randomly mixed into the past, and the notebook explicitly verifies that the three periods do not overlap. The final evaluation is therefore conducted on observations that were not used in model estimation.

Second, simple and deliberately weak baselines challenge the more sophisticated models. Logistic regression, random forest, and the neural network do not receive credit merely for producing predictions; they must demonstrate improvement over naive, random, passive, technical, and autoregressive alternatives.

Third, the notebook examines probability dispersion, calibration-sensitive loss, and the proportion of high-confidence forecasts. This helps identify models that appear accurate only because they remain close to the unconditional probability of an upward movement.

Fourth, strategies are exposed to different synthetic regimes and to economically relevant implementation conditions, including turnover, transaction costs, volatility differences, position limits, sector limits, and gross-exposure constraints.

However, the current notebook does **not yet complete a full model-by-model robustness campaign**. The downstream stress stage should rerun the models under:

- Alternative synthetic-market seeds
- Different train, validation, and test windows
- Walk-forward and nested validation
- Feature perturbations and feature removal
- Hyperparameter perturbations
- Alternative signal thresholds
- Higher transaction costs and wider spreads
- Liquidity shocks
- More severe or longer crisis regimes
- Missing, stale, or corrupted data
- Independent risk and audit challenge

The candidate remains experimental until it survives those additional tests.

### F. How is the success of each model measured?

Success is measured at two separate levels.

#### Predictive success

| Metric | Interpretation |
|---|---|
| **Accuracy** | Percentage of next-day directions classified correctly. |
| **Balanced accuracy** | Average accuracy across the positive and negative classes, preventing the more frequent class from dominating the result. |
| **Brier score** | Mean squared error of the predicted probabilities. Lower values indicate better probabilistic accuracy. |
| **Probability mean** | Shows the model’s average directional bias. |
| **Probability standard deviation** | Measures how much the model differentiates among observations. |
| **High-confidence share** | Percentage of probabilities above 0.60 or below 0.40. |

#### Economic success

| Metric | Interpretation |
|---|---|
| **Final equity** | Ending value of one initial unit of capital. |
| **Annualized return** | Average daily net return converted to an annual rate. |
| **Annualized volatility** | Annualized variability of daily net returns. |
| **Sharpe ratio** | Annualized return divided by annualized volatility. |
| **Maximum drawdown** | Largest decline from a previous portfolio peak. |
| **Turnover** | Average amount of portfolio reallocation. |
| **Constraint breaches** | Number of dates on which gross or concentration limits were violated. |
| **Regime and sector attribution** | Identifies where returns, risks, and trading activity originated. |

A model is not successful merely because it has the highest accuracy. It must produce useful probabilities, support an economically coherent signal, survive implementation costs, respect portfolio constraints, and remain stable under additional challenge.

All predictive models receive:

1. The same six input features.
2. The same next-day target.
3. The same chronological training and testing periods.
4. The same stock universe.

The economic scorecard is separate. It compares the 20 strategy–portfolio combinations using return, volatility, Sharpe ratio, drawdown, turnover, and constraint compliance. In the validated run, the leading research candidate is:

Its approximate results are:

- Maximum drawdown: **−1.3%**

This does not mean that the random forest is universally superior to logistic regression. Logistic regression leads the predictive comparison, while the random forest currently powers the thresholded ML strategy. The notebook deliberately keeps predictive leadership and economic leadership as separate conclusions.

### H. What is the final lesson the user should retain?

The main lesson is that **a predictive model is not yet an investment strategy, and an investment strategy is not yet a governed portfolio**.

Information must pass through a complete institutional chain:

$
\text{Data}
\rightarrow
\text{Features}
\rightarrow
\text{Predictions}
\rightarrow
\text{Signals}
\rightarrow
\text{Portfolio Weights}
\rightarrow
\text{Costs and Constraints}
\rightarrow
\text{Net Performance}
\rightarrow
\text{Audit Evidence}.
$

NB01 teaches us that predictive quality and economic value are different questions. Portfolio construction can materially change the expression of the same signal. Transaction costs, turnover, exposure limits, and drawdowns are part of the strategy itself—not adjustments added after the modelling work is complete.

The notebook’s leading result is therefore a **research candidate**, not an investment recommendation. Its official status remains:
