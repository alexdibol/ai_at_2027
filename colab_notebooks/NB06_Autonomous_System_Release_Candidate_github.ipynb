# NB06 — Autonomous System Release Candidate

## Introduction

NB06 is the point at which the course’s governed research components become an executable autonomous system. The previous notebooks constructed the institution in deliberate layers. NB00A established the research-only protocol and its authority boundary. NB00 created a reproducible synthetic-equity environment. NB00B demonstrated an end-to-end KNN research slice. NB01 expanded that slice into a comparative laboratory of models, signals, strategies, and portfolios. NB02 added simulated execution, independent risk, governance gates, provenance, and audit. NB03 converted validated functions into 35 tools, nine skills, and 13 bounded agents. NB04 assembled those roles into four institutional constellations. NB05 introduced a governed meta-agent that could express a mission as an observable dependency plan. NB06 now integrates those parts into a closed control loop that receives an objective, interprets it, constructs and labels tasks, discovers eligible routes, executes the plan, validates outputs, monitors progress, recovers from bounded failures, and closes with a hash-linked audit record.

This distinction matters because a catalog of components is not yet an autonomous architecture. A static plan can demonstrate a desirable route, and a table of expected outcomes can document policy intent, but neither proves that the route was selected from the content of a new task. Autonomy begins when the system can transform an admitted objective into runtime decisions without a human manually assigning every intermediate step. In NB06, the original Day 9 records are preserved exactly as regression fixtures, but the enhanced notebook no longer treats those records as sufficient evidence. It adds executable functions that generate the task graph, labels, route assignments, checkpoints, traces, mission results, monitor state, adversarial outcomes, and release decision during notebook execution.

The resulting autonomy is real but deliberately narrow. It is not general intelligence, unsupervised production operation, or live trading. It is bounded research autonomy inside a deterministic synthetic-equity sandbox. The system may autonomously decide which approved research procedure should handle a task, but it may choose only among validated routes already recorded in its in-memory registry. It may recover from one modeled failure, but it may not expand its retry allowance. It may complete a research assessment, but it may not promote a strategy. It may produce a simulated portfolio, but it may not connect to a broker, transmit an order, deploy code, or create any external market effect. Autonomy describes control over an allowed workflow; it does not imply ownership of the institution’s material decisions.

Every mission begins as a structured objective. The admission layer requires a mission identifier, objective, asset class, scope, evidence, requested output, and decision owner. These fields answer the preliminary questions needed for safe interpretation: what is being requested, where it applies, which evidence supports it, what result is expected, and who remains accountable. The interpreter normalizes the objective and computes a semantic hash. Missing fields cause escalation. A non-synthetic asset class is denied. Live-order or deployment language is denied. Concealment or audit-bypass intent is denied. A request for strategy promotion, a risk override, deployment, or live execution is classified as a material decision and transferred to the human governance owner.

An admitted objective then reaches the planner. Unlike the static example in NB05, the enhanced planner constructs a dependency graph according to the mission’s content. Every mission needs research evidence, independent risk challenge, and audit closure. Strategy evaluation is inserted when the objective refers to strategies, momentum, signals, backtesting, or portfolios. Portfolio construction is inserted only when the objective requires positions, weights, allocation, or a portfolio. The raw tasks intentionally contain neither a task label nor an owner. This makes the subsequent decisions observable: the system cannot claim dynamic classification if the tasks arrive pre-labeled or pre-routed.

Task labeling uses a controlled ontology. Each label declares its keywords, required capabilities, output type, estimated cost, tool calls, and whether institutional independence is required. The labeler scores a raw task against this vocabulary and returns the best supported label, a confidence value, and a label hash. Unknown task types fail closed by escalating rather than falling through to a generic handler. This is a deterministic educational classifier rather than a learned language model, but it demonstrates the architecture required for labeling: an explicit ontology, reproducible decision procedure, confidence, traceability, and a refusal path.

Route selection is separate from labeling. A task can be correctly described yet still lack an authorized route. The selector queries the current capability registry for validated constellations that support the synthetic-equity scope, contain all required capabilities, carry no live authority, and satisfy separation of duties. It records the eligibility or rejection reason for every candidate. If several routes qualify, it selects the least-privileged match by minimizing surplus capabilities and then applying a declared priority. If no route qualifies, the task is denied. This answers the central routing question from NB05: the route is not chosen because the task arrived with an owner. It is computed from the intersection of the task label, required capabilities, scope, registry status, authority, independence requirement, and resource constraints.

The system then validates the dependency graph, totals the planned cost and tool calls, and binds the result to a plan hash. A mission cannot enter execution if its plan exceeds the sandbox budget. During execution, each task must have satisfied dependencies and sufficient remaining steps, calls, cost units, and artifact capacity. The controller produces a typed research artifact, validates its schema, records its producer and parent hashes, consumes the authorized budget, appends a hash-linked audit event, emits a trace record, and creates a recoverable checkpoint. These operations turn the plan into observable state rather than a narrative assertion.

The state machine governs the entire lifecycle. Twelve states represent creation, validation, planning, readiness, execution, pausing, recovery, escalation, completion, stopping, abortion, and failure. A transition function checks every requested movement against the allowlist and records both accepted and denied transitions. Terminal states cannot resume. This prevents a convenient procedural shortcut from reviving a mission that has already completed, failed, stopped, or been aborted. The audit chain links each event to its predecessor, while checkpoints bind the mission state, remaining budget, artifact lineage, and audit head to a content hash.

Recovery is bounded and visible. The failure-recovery mission deliberately injects a schema problem during strategy evaluation. The invalid output is rejected, the state moves into recovery, a checkpoint is written, a small recovery cost is charged, and the task is retried once through the same authorized route. The corrected output must pass the ordinary contract before execution continues. Recovery therefore means that policy-controlled remediation succeeded; it does not erase the exception or pretend that the first output was valid.

Five runtime missions exercise the integrated system. A single-sector momentum question demonstrates a four-stage research route. A multi-sector and multi-regime question tests adaptation without leaving the synthetic-equity asset class. A portfolio request adds a fifth task and therefore proves that the plan changes with the objective. A schema-failure mission tests recovery. A strategy-promotion request escalates before planning because the requested outcome belongs to a human decision owner. The different task sequences produce multiple plan signatures, directly demonstrating that the enhanced notebook is not replaying one hard-coded route for every input.

Ten additional adversarial tests probe non-synthetic scope, budget exhaustion, human escalation, model failure, schema failure, live orders, terminal-state resumption, audit-logger failure, checkpoint tampering, and self-promotion by the release candidate. The test outcomes are computed by invoking the control functions, not copied from an expected-results table. Every test must match its declared outcome before the release candidate can pass. Monitoring is similarly derived from executed missions: heartbeat, audit integrity, budget utilization, artifact counts, completed missions, recovered exceptions, unresolved critical exceptions, and live-order attempts are calculated from runtime records.

The release decision is intentionally qualified. A pass establishes that the notebook contains a coherent bounded research control plane and that its deterministic tests behave as specified. It does not establish profitable investment performance, secure distributed infrastructure, real-time service discovery, production-grade identity and access management, durable checkpoint storage, cryptographic signing, or regulatory readiness. The capability registry remains an in-memory educational object, and task interpretation remains a controlled rules engine. These limitations are preserved in the handoff rather than hidden behind the word autonomous.

NB06 therefore completes the transition from governed components to bounded autonomous research. NB07 will use this release candidate as the control layer for a research-to-LEAN bridge and experiment registry. That next stage can translate approved research artifacts toward platform-compatible experiments, but it must inherit the same authority constraints: external writes require approval, deployment is not automatic, and live trading remains denied. The institutional doctrine is consistent across the course: the system may increasingly interpret, plan, route, execute, monitor, recover, and audit within its approved domain, while humans retain authority over promotion, exceptions, deployment, and every action with a material external effect.

## Workflow: how NB06 converts an objective into bounded autonomous research

The enhanced notebook runs a complete control loop. Raw objectives do not arrive with owners. The system interprets their semantics, decomposes them into unlabeled tasks, assigns controlled labels, discovers current capabilities, selects the least-privileged authorized route, validates dependencies and budgets, executes typed research procedures, checks every output, writes artifacts and checkpoints, monitors the mission, and then completes, escalates, denies, stops, or fails according to policy.

| Stage | Runtime decision | Evidence created |
|---|---|---|
| 1. Admit the mission | Are all semantics present, is the scope synthetic, and is the requested authority permitted? | Admission decision and semantic hash |
| 2. Decompose | Which tasks and dependencies are implied by this objective? | Raw task graph with no labels or owners |
| 3. Label | Which controlled task type best describes each task? | Label, confidence, required capabilities, output contract, label hash |
| 4. Discover | Which validated routes currently cover the scope and capabilities? | Registry snapshot and discovery hash |
| 5. Route | Which eligible route has the least surplus authority and preserves independence? | Selected constellation plus reasons for every candidate |
| 6. Validate the plan | Are dependencies acyclic and are total calls and costs within budget? | Plan, resource totals, and plan hash |
| 7. Execute | Are dependencies complete and resources still available for the next task? | Typed artifact, trace record, budget consumption, and audit event |
| 8. Validate and recover | Does the output satisfy its schema, and may one bounded retry correct a failure? | Rejection event, recovery checkpoint, retry result, and preserved exception |
| 9. Monitor | Are audit integrity, budgets, artifacts, exceptions, and prohibited attempts observable? | Monitor derived from mission records |
| 10. Close | Has the mission completed as research-only, been escalated, or stopped at a hard boundary? | Final state, open questions, promotion denial, and release evidence |

The five mission examples produce two distinct plan shapes. Research and strategy objectives route through Research, Strategy, Risk Governance, and Audit. A portfolio objective inserts Portfolio between Strategy and Risk Governance. A promotion request does not receive a plan at all; it escalates at admission. This variation is the notebook’s direct evidence that tasks are labeled and routes are selected from runtime inputs rather than simply replayed from one fixed sequence.

## 1. Integration mission and deterministic primitives

Day 9 begins by preserving the original release-candidate declaration and adding the canonical serialization, hashing, and audit primitives required by the runtime. Every later plan, label, artifact, checkpoint, test, and release record can therefore be tied to a reproducible digest. Audit events form a chain: each event contains the previous event’s hash, making deletion, reordering, or silent modification detectable within the notebook’s educational model.

```python
import json, hashlib
print('Day 9 sandbox ready')

# --- Executable Day 9 extension ---
from copy import deepcopy
from collections import Counter
import re


def canonical_json(value):
    return json.dumps(value, sort_keys=True, separators=(",", ":"), ensure_ascii=False)


def stable_hash(value):
    return hashlib.sha256(canonical_json(value).encode("utf-8")).hexdigest()


def append_audit(context, action, payload):
    previous_hash = context["audit"][-1]["event_hash"] if context["audit"] else "GENESIS"
    event = {
        "sequence": len(context["audit"]) + 1,
        "mission_id": context["spec"].get("mission_id", "UNKNOWN"),
        "state": context["state"],
        "action": action,
        "payload_hash": stable_hash(payload),
        "previous_hash": previous_hash,
    }
    event["event_hash"] = stable_hash(event)
    context["audit"].append(event)
    return event


def verify_audit_chain(events):
    previous_hash = "GENESIS"
    for expected_sequence, event in enumerate(events, start=1):
        body = {key: value for key, value in event.items() if key != "event_hash"}
        if event["sequence"] != expected_sequence:
            return False
        if event["previous_hash"] != previous_hash:
            return False
        if event["event_hash"] != stable_hash(body):
            return False
        previous_hash = event["event_hash"]
    return True


print("deterministic hashing and hash-linked audit primitives PASS")
```

## 2. Executable fail-closed state machine

The original twelve-state transition table becomes an enforced lifecycle. The transition function applies only listed movements and records an explicit denial for every invalid request. Completion, stopping, abortion, and failure are terminal. This matters because autonomy requires a system to know not only what action comes next, but also when no further action is allowed.

```python
STATES=["CREATED", "VALIDATING", "PLANNING", "READY", "RUNNING", "PAUSED", "RECOVERING", "ESCALATED", "COMPLETED", "STOPPED", "ABORTED", "FAILED"]
TRANSITIONS=json.loads("{\"CREATED\": [\"VALIDATING\", \"ABORTED\"], \"VALIDATING\": [\"PLANNING\", \"ESCALATED\", \"ABORTED\", \"FAILED\"], \"PLANNING\": [\"READY\", \"ESCALATED\", \"ABORTED\", \"FAILED\"], \"READY\": [\"RUNNING\", \"STOPPED\", \"ABORTED\"], \"RUNNING\": [\"PAUSED\", \"RECOVERING\", \"ESCALATED\", \"COMPLETED\", \"STOPPED\", \"ABORTED\", \"FAILED\"], \"PAUSED\": [\"RUNNING\", \"STOPPED\", \"ABORTED\"], \"RECOVERING\": [\"RUNNING\", \"ESCALATED\", \"ABORTED\", \"FAILED\"], \"ESCALATED\": [\"RUNNING\", \"STOPPED\", \"ABORTED\"], \"COMPLETED\": [], \"STOPPED\": [], \"ABORTED\": [], \"FAILED\": []}")
assert len(STATES)==12
print('state machine PASS')

# --- Executable Day 9 extension ---
TERMINAL_STATES = {"COMPLETED", "STOPPED", "ABORTED", "FAILED"}


def transition(context, target_state, event):
    current_state = context["state"]
    allowed = target_state in TRANSITIONS.get(current_state, [])
    if not allowed:
        append_audit(context, "TRANSITION_DENIED", {
            "from": current_state,
            "to": target_state,
            "event": event,
        })
        return {
            "status": "DENIED",
            "from": current_state,
            "to": current_state,
            "reason": "INVALID_STATE_TRANSITION",
        }
    context["state"] = target_state
    append_audit(context, "STATE_TRANSITION", {
        "from": current_state,
        "to": target_state,
        "event": event,
    })
    return {
        "status": "APPLIED",
        "from": current_state,
        "to": target_state,
        "reason": event,
    }


assert set(TRANSITIONS) == set(STATES)
assert all(target in STATES for targets in TRANSITIONS.values() for target in targets)
assert all(TRANSITIONS[state] == [] for state in TERMINAL_STATES)
print("executable fail-closed state machine PASS")
```

## 4. Sandbox, objective interpretation, and mission context

The interpreter is the first runtime gate. It requires complete semantics, rejects non-synthetic and live-action requests, refuses concealment intent, and escalates material decisions. An admitted mission receives isolated working context, explicit budgets, an empty audit chain, and no inherited task ownership. Network, subprocess, brokerage, credentials, and live orders remain denied exactly as in the original notebook.

```python
SANDBOX=json.loads("{\"network\": \"deny\", \"subprocess\": \"deny\", \"brokerage\": \"deny\", \"live_orders\": \"deny\", \"credentials\": \"none\", \"filesystem\": \"run_artifact_directory_only\", \"seed\": 20260818, \"budgets\": {\"max_runtime_seconds\": 120, \"max_memory_mb\": 512, \"max_api_calls\": 0, \"max_tool_calls\": 60, \"max_steps\": 20, \"max_recursion_depth\": 3, \"max_artifacts\": 50, \"max_cost_units\": 120}}")
LIFECYCLE=json.loads("[{\"from\": \"READY\", \"event\": \"start\", \"status\": \"APPLIED\", \"to\": \"RUNNING\"}, {\"from\": \"RUNNING\", \"event\": \"pause\", \"status\": \"APPLIED\", \"to\": \"PAUSED\"}, {\"from\": \"PAUSED\", \"event\": \"resume\", \"status\": \"APPLIED\", \"to\": \"RUNNING\"}, {\"from\": \"RUNNING\", \"event\": \"stop\", \"status\": \"APPLIED\", \"to\": \"STOPPED\"}, {\"from\": \"READY\", \"event\": \"abort\", \"status\": \"APPLIED\", \"to\": \"ABORTED\"}, {\"from\": \"COMPLETED\", \"event\": \"resume\", \"status\": \"DENIED\", \"to\": \"COMPLETED\"}]")
assert SANDBOX['live_orders']=='deny' and LIFECYCLE[-1]['status']=='DENIED'
print('sandbox and lifecycle PASS')

# --- Executable Day 9 extension ---
REFERENCE_LIFECYCLE = deepcopy(LIFECYCLE)

REQUIRED_OBJECTIVE_FIELDS = {
    "mission_id", "objective", "asset_class", "scope",
    "evidence", "requested_output", "decision_owner"
}

LIVE_ACTION_TERMS = {
    "live order", "send order", "execute trade", "brokerage",
    "connect broker", "authorize deployment", "deploy to production"
}
DECEPTIVE_TERMS = {"hide evidence", "conceal", "bypass audit", "suppress dissent"}
MATERIAL_OUTPUTS = {"strategy_promotion", "risk_override", "deployment", "live_execution"}


def interpret_objective(spec):
    missing = sorted(REQUIRED_OBJECTIVE_FIELDS - set(spec))
    if missing:
        return {
            "decision": "ESCALATE",
            "reason": "SEMANTIC_INSUFFICIENCY",
            "missing": missing,
        }
    normalized_text = " ".join([
        str(spec.get("objective", "")),
        str(spec.get("scope", "")),
        " ".join(spec.get("side_effects", [])),
    ]).lower()
    if spec["asset_class"] != "synthetic_equities":
        return {"decision": "DENY", "reason": "NON_SYNTHETIC_SCOPE"}
    if any(term in normalized_text for term in LIVE_ACTION_TERMS):
        return {"decision": "DENY", "reason": "LIVE_AUTHORITY_PROHIBITED"}
    if any(term in normalized_text for term in DECEPTIVE_TERMS):
        return {"decision": "DENY", "reason": "DECEPTIVE_OR_CONCEALMENT_INTENT"}
    if spec["requested_output"] in MATERIAL_OUTPUTS or "promote" in normalized_text:
        return {
            "decision": "REQUIRE_HUMAN_APPROVAL",
            "reason": "MATERIAL_DECISION",
            "checkpoint": "HUMAN_GOVERNANCE_OWNER",
        }
    parsed = {
        **deepcopy(spec),
        "normalized_objective": re.sub(r"\s+", " ", spec["objective"].strip().lower()),
        "semantic_hash": stable_hash(spec),
    }
    return {"decision": "ADMIT", "reason": "ALL_ADMISSION_GATES_PASSED", "parsed": parsed}


def create_context(spec, budget_override=None):
    budget = {
        "steps": SANDBOX["budgets"]["max_steps"],
        "tool_calls": SANDBOX["budgets"]["max_tool_calls"],
        "cost_units": SANDBOX["budgets"]["max_cost_units"],
        "artifacts": SANDBOX["budgets"]["max_artifacts"],
        "retries_per_task": 1,
    }
    if budget_override:
        budget.update(budget_override)
    return {
        "spec": deepcopy(spec),
        "state": "CREATED",
        "audit": [],
        "trace": [],
        "artifacts": [],
        "checkpoints": [],
        "completed_tasks": set(),
        "budget_initial": deepcopy(budget),
        "budget_remaining": deepcopy(budget),
        "retry_count": {},
        "recovered": False,
        "failure_injected": False,
    }


assert SANDBOX["network"] == "deny"
assert SANDBOX["subprocess"] == "deny"
assert SANDBOX["brokerage"] == "deny"
assert SANDBOX["live_orders"] == "deny"
print("dynamic admission, mission context, and sandbox boundaries PASS")
```

## 5. Dynamic decomposition, labeling, routing, and checkpoints

The planner derives a task graph from the objective. The labeler then maps every raw task to the controlled ontology, and the router evaluates every registered constellation against capabilities, scope, lifecycle status, live authority, and independence. Selection follows a least-privilege rule. Checkpoints bind state, remaining resources, artifact hashes, and the current audit head, enabling later integrity verification.

```python
CHECKPOINTS=json.loads("[{\"checkpoint_id\": \"D9_M04_FAILURE_RECOVERY_CP_03\", \"mission_id\": \"D9_M04_FAILURE_RECOVERY\", \"state\": \"RUNNING\", \"step\": 3, \"budget_remaining\": {\"steps\": 17, \"cost_units\": 92}, \"artifact_hashes\": [\"1766124aa582560388f02962c035a0fbc148757d6469c236545ccea1c35e38e2\"], \"audit_head\": \"audit_pre\", \"recoverable\": true, \"checkpoint_hash\": \"2bfc126ab361b4f880ba62722a5db254e97c4e2da0c24a36eca57144d20cf9a3\"}, {\"checkpoint_id\": \"D9_M04_FAILURE_RECOVERY_CP_04\", \"mission_id\": \"D9_M04_FAILURE_RECOVERY\", \"state\": \"RUNNING\", \"step\": 4, \"budget_remaining\": {\"steps\": 16, \"cost_units\": 88}, \"artifact_hashes\": [\"1766124aa582560388f02962c035a0fbc148757d6469c236545ccea1c35e38e2\", \"e59afe2291ec6b68bdeb50656a3c728b431ba040524a4dd2bab6b336f439af15\"], \"audit_head\": \"audit_post\", \"recoverable\": true, \"checkpoint_hash\": \"d5777ce26f41ad93594c60f86297569cc94ff4df52ea6f5ec9f2c449f974e04f\"}]")
assert CHECKPOINTS[1]['recoverable']
print('checkpoint and provenance PASS')

# --- Executable Day 9 extension ---
REFERENCE_CHECKPOINTS = deepcopy(CHECKPOINTS)


def decompose_mission(parsed):
    text = parsed["normalized_objective"]
    tasks = []

    def add(description, dependencies):
        task_id = f"T{len(tasks) + 1}"
        tasks.append({
            "task_id": task_id,
            "description": description,
            "depends_on": list(dependencies),
        })
        return task_id

    research_id = add(
        "Prepare validated point-in-time data, model diagnostics, and research evidence.",
        [],
    )
    latest_dependencies = [research_id]
    strategy_terms = {"strategy", "momentum", "signal", "backtest", "portfolio"}
    need_strategy = any(term in text for term in strategy_terms)
    need_portfolio = any(term in text for term in {"portfolio", "allocation", "positions", "weights"})

    if need_strategy:
        strategy_id = add(
            "Design and backtest an experimental strategy using the research evidence.",
            latest_dependencies,
        )
        latest_dependencies = [strategy_id]

    if need_portfolio:
        portfolio_id = add(
            "Construct a constrained simulated portfolio from the experimental strategy.",
            latest_dependencies,
        )
        latest_dependencies = [portfolio_id]

    upstream_ids = [task["task_id"] for task in tasks]
    risk_id = add(
        "Independently review risk, stress behavior, limits, and governance evidence.",
        upstream_ids,
    )
    add(
        "Verify provenance and assemble the final audit bundle with unresolved questions.",
        [risk_id],
    )
    assert all("label" not in task and "owner" not in task for task in tasks)
    return tasks


def label_task(task):
    text = task["description"].lower()
    ordered_labels = [
        "audit_bundle", "risk_challenge", "portfolio_construction",
        "strategy_evaluation", "research_evidence"
    ]
    scored = []
    for label in ordered_labels:
        definition = TASK_ONTOLOGY[label]
        score = sum(keyword in text for keyword in definition["keywords"])
        scored.append((score, -ordered_labels.index(label), label))
    best_score, _, best_label = max(scored)
    if best_score == 0:
        return {
            "decision": "ESCALATE",
            "reason": "UNKNOWN_TASK_TYPE",
            "confidence": 0.0,
        }
    definition = TASK_ONTOLOGY[best_label]
    confidence = min(0.99, 0.70 + 0.08 * best_score)
    return {
        "decision": "LABELED",
        "label": best_label,
        "required_capabilities": sorted(definition["required_capabilities"]),
        "output_type": definition["output_type"],
        "cost_units": definition["cost_units"],
        "tool_calls": definition["tool_calls"],
        "independence_required": definition["independence_required"],
        "confidence": confidence,
        "label_hash": stable_hash({"task": task, "label": best_label}),
    }


def select_route(labeled_task, scope, producer_routes):
    required = set(labeled_task["required_capabilities"])
    discovery = discover_capabilities(scope, required)
    considered = {}
    eligible = []
    for route_id, entry in CAPABILITY_REGISTRY.items():
        reasons = []
        if entry["status"] != "validated":
            reasons.append("NOT_VALIDATED")
        if entry["live_authority"]:
            reasons.append("LIVE_AUTHORITY_PRESENT")
        if scope not in entry["scopes"]:
            reasons.append("SCOPE_MISMATCH")
        if not required.issubset(entry["capabilities"]):
            reasons.append("INCOMPLETE_CAPABILITY_COVERAGE")
        if labeled_task["independence_required"] and route_id in producer_routes:
            reasons.append("INSUFFICIENT_INDEPENDENCE")
        considered[route_id] = "ELIGIBLE" if not reasons else ",".join(reasons)
        if not reasons:
            surplus = len(entry["capabilities"] - required)
            eligible.append((surplus, entry["priority"], route_id))
    if not eligible:
        return {
            "decision": "DENY",
            "reason": "NO_AUTHORIZED_ROUTE",
            "considered": considered,
            "discovery_hash": discovery["registry_hash"],
        }
    eligible.sort()
    selected = eligible[0][2]
    return {
        "decision": "ROUTE",
        "selected_route": selected,
        "reason": "LEAST_PRIVILEGE_CAPABILITY_MATCH",
        "considered": considered,
        "discovery_hash": discovery["registry_hash"],
    }


def build_plan(parsed):
    raw_tasks = decompose_mission(parsed)
    plan = []
    producer_routes = set()
    for raw_task in raw_tasks:
        label = label_task(raw_task)
        if label["decision"] != "LABELED":
            return {"status": "ESCALATE", "reason": label["reason"], "raw_tasks": raw_tasks}
        route = select_route(label, parsed["asset_class"], producer_routes)
        if route["decision"] != "ROUTE":
            return {"status": "DENY", "reason": route["reason"], "raw_tasks": raw_tasks}
        task = {
            **raw_task,
            **{key: value for key, value in label.items() if key != "decision"},
            "route": route["selected_route"],
            "route_reason": route["reason"],
            "route_candidates": route["considered"],
            "discovery_hash": route["discovery_hash"],
        }
        plan.append(task)
        if not task["independence_required"]:
            producer_routes.add(task["route"])

    known = set()
    for task in plan:
        if not set(task["depends_on"]).issubset(known):
            return {"status": "DENY", "reason": "INVALID_DEPENDENCY_GRAPH", "raw_tasks": raw_tasks}
        known.add(task["task_id"])
    plan_payload = {
        "mission_id": parsed["mission_id"],
        "tasks": plan,
        "total_cost_units": sum(task["cost_units"] for task in plan),
        "total_tool_calls": sum(task["tool_calls"] for task in plan),
    }
    plan_payload["plan_hash"] = stable_hash(plan_payload)
    return {"status": "PLANNED", "raw_tasks": raw_tasks, **plan_payload}


def make_checkpoint(context, step, recoverable=True):
    payload = {
        "checkpoint_id": f"{context['spec'].get('mission_id', 'UNKNOWN')}_CP_{step:02d}",
        "mission_id": context["spec"].get("mission_id", "UNKNOWN"),
        "state": context["state"],
        "step": step,
        "budget_remaining": deepcopy(context["budget_remaining"]),
        "artifact_hashes": [artifact["artifact_hash"] for artifact in context["artifacts"]],
        "audit_head": context["audit"][-1]["event_hash"] if context["audit"] else "GENESIS",
        "recoverable": recoverable,
    }
    payload["checkpoint_hash"] = stable_hash(payload)
    context["checkpoints"].append(payload)
    return payload


def verify_checkpoint(checkpoint):
    body = {key: value for key, value in checkpoint.items() if key != "checkpoint_hash"}
    return checkpoint.get("checkpoint_hash") == stable_hash(body)


print("dynamic decomposition, task labeling, route selection, and checkpointing PASS")
```

Dynamic planning means that NB06 does not execute the same predetermined workflow for every request. Instead, it examines the objective, determines which tasks are required, labels those tasks, identifies eligible capabilities, and selects an authorized route during execution.

Scope adaptation means that this flexibility remains inside the system’s approved boundary. NB06 may vary the sectors, market regimes, analytical procedures, and plan structure used in an experiment, but it cannot silently introduce a new asset class or acquire additional authority.

### Why the multi-market requirement is adapted

A literal interpretation of “multi-market” could introduce currencies, commodities, credit instruments, futures, or derivatives. However, these markets require different:

- Data structures
- Trading calendars
- Valuation methods
- Execution models
- Liquidity assumptions
- Margin and collateral rules
- Risk controls

NB06 has validated only a synthetic-equity research environment. It therefore translates the broader requirement into multiple equity sectors and multiple market regimes.

For example, it may study:

- Technology, financial, industrial, healthcare, and consumer sectors
- Calm, trending, crisis, and liquidity-stressed regimes

This produces substantial analytical variation while preserving the approved synthetic-equity boundary. The system becomes more flexible without claiming capabilities that have not been validated.

|---|---|---|

### 2. Labeling classifies each task

The labeler compares each task description with the controlled task ontology. It assigns the following labels:

| Task | Computed label | Required output |
|---|---|---|
| T1 | `research_evidence` | Validated research evidence |
| T2 | `strategy_evaluation` | Experimental strategy proposal |
| T3 | `portfolio_construction` | Constrained portfolio report |

- Expected output type
- Estimated cost
- Estimated tool calls
- Independence requirement
- Classification confidence
- Label hash

If no ontology category adequately describes a task, the system does not guess. It escalates the unknown task for review.

### 3. Discovery identifies eligible routes

The discovery function examines the current capability registry. A constellation is eligible only if it:

1. Has validated status.
2. Supports the synthetic-equity scope.
3. Contains all capabilities required by the task.
4. Possesses no live-trading authority.
5. Satisfies any applicable independence requirement.

Discovery therefore determines which routes are currently available. It does not automatically authorize every registered component.

The registry snapshot receives a cryptographic hash, connecting the routing decision to the exact capability configuration available when the plan was created.

### 4. Routing selects the least-privileged match

The router evaluates every candidate and records why it is eligible or rejected. Among the eligible candidates, it selects the route with the fewest unnecessary capabilities.

The resulting plan is:

| Task | Selected route | Routing reason |
|---|---|---|
| T1 | `research_team` | Validated research capability match |
| T2 | `strategy_team` | Validated strategy and backtesting match |
| T3 | `portfolio_team` | Validated portfolio-construction match |
| T4 | `risk_governance_team` | Independent risk and governance capability |
| T5 | `risk_governance_team` | Audit and provenance capability |

T4 cannot be assigned to the Strategy or Portfolio Team because independent review must remain separate from the teams that produced the proposal. This is how NB06 enforces separation of duties.

The governing rule is:

$
\text{Effective Route}
=
\text{Task Label}
\cap
\text{Required Capabilities}
\cap
\text{Validated Registry}
\cap
\text{Scope}
\cap
\text{Independence}
\cap
\text{Budget}
$

If this intersection produces no eligible route, the task is denied.

# Frequently Asked Questions — NB06

## 1. What does NB06 add to the course architecture?

NB06 integrates the earlier registries, agents, constellations, governance rules, and meta-agent concepts into one executable research control loop. It can admit a structured objective, create tasks, label them, discover eligible capabilities, choose routes, execute typed procedures, validate outputs, manage budgets, recover once from a modeled failure, checkpoint state, derive monitoring, and close with auditable evidence.

## 2. Is NB06 genuinely autonomous?

Yes, within a narrow and explicitly governed sense. Its runtime functions make intermediate decisions from each objective rather than requiring a human to assign every task and route. The autonomy is deterministic, synthetic, and research-only. It does not include unrestricted learning, open internet use, production deployment, broker connectivity, or authority over material decisions.

## 3. Why was NB05 not yet a complete autonomous architecture?

NB05 specified the planner, budgets, routes, and expected outcomes, but its principal five-task plan and scenario results were static records. That was enough to explain and validate the orchestration contract, but not enough to demonstrate that new inputs could be interpreted and routed at runtime. NB06 turns those specifications into callable decision functions and executed mission traces.

## 4. Are the original NB06 cells discarded?

No. Every original code cell is preserved verbatim at the beginning of its corresponding enhanced code cell. Those objects serve as regression fixtures: they document the authoritative Day 9 expectations. The new runtime layer must reproduce and strengthen the intended behavior rather than silently replacing the original evidence.

## 5. Does a task arrive with a label?

No. The decomposer creates raw tasks with identifiers, descriptions, and dependencies only. Assertions verify that the raw tasks contain neither a label nor an owner. The labeler subsequently classifies each task using the controlled ontology. This separation makes the classification decision observable and prevents the system from presenting a pre-labeled example as dynamic interpretation.

## 6. How is a task labeled?

Each ontology entry declares keywords, required capabilities, an output type, estimated cost, estimated tool calls, and whether independence is required. The labeler scores the task description against the vocabulary, applies a deterministic priority for ties, and returns the best supported type with confidence and a label hash. If no type receives support, the task escalates as unknown.

## 7. Is keyword labeling sufficient for production autonomy?

No. It is a transparent educational implementation that proves the interface and control requirements. A production classifier might combine schemas, rules, statistical models, and language models, but it would still need controlled labels, calibrated confidence, out-of-distribution detection, evaluation sets, versioning, provenance, human escalation, and protection against prompt injection and adversarial phrasing.

## 8. How is the route selected?

The router takes the computed task label and its required capabilities, then evaluates the current registry. A route must be validated, support the synthetic-equity scope, contain every required capability, carry no live authority, and preserve independence when the task is a risk or audit function. The least-privileged eligible route is chosen by minimizing surplus capabilities and applying a stable priority.

## 9. Is routing hard-coded?

The registry and routing policy are declared in code, as policies must be, but the selected route is computed at runtime. The mission input does not contain an owner. Each candidate receives an eligibility or rejection reason, and the selected constellation follows from the task requirements and current registry. The notebook therefore distinguishes configured policy from a preassigned route.

## 10. What happens if no route qualifies?

The route decision becomes DENY with reason NO_AUTHORIZED_ROUTE. The record retains every candidate and its rejection reasons. The planner does not choose an approximate team, invent a missing capability, remove the independence requirement, or broaden the scope. A new route would require an explicit registry and governance change outside the active mission.

## 11. What is least-privilege routing?

When multiple routes satisfy all mandatory conditions, the system prefers the one with the fewest capabilities beyond those required by the task. This reduces the authority surface exposed to the mission. Stable priority breaks ties. Least privilege is not merely a descriptive principle here; it is part of the route-selection algorithm and is recorded as the decision reason.

## 12. How is separation of duties enforced?

The planner tracks routes that produced upstream research, strategy, or portfolio artifacts. Tasks marked independent, such as risk challenge and audit closure, cannot be assigned to those producer routes. The current registry naturally sends independent review to Risk Governance, but the explicit check prevents a future registry expansion from quietly allowing a producer to validate itself.

## 13. How does objective interpretation work?

The interpreter verifies seven required fields, normalizes the objective, and evaluates prohibited or material intent. Missing semantics escalate. A non-synthetic asset class, live action, deployment language, or concealment intent is denied. Strategy promotion and other material outputs require human approval. Only an admitted objective receives a semantic hash and proceeds into planning.

## 14. Why does a promotion request stop before planning?

Promotion is a material governance decision, not a research procedure. Decomposing it into executable research tasks could misleadingly imply that successful analysis would authorize the requested status change. NB06 instead recognizes the authority boundary at admission, enters the escalated state, records the reason, and transfers the decision to the human governance owner.

## 15. How does the task graph change with the objective?

Research evidence, independent challenge, and audit closure are mandatory for admitted research missions. Strategy evaluation appears when the objective concerns strategies, signals, momentum, backtests, or portfolios. Portfolio construction appears only when positions, weights, allocation, or a portfolio are requested. The mission suite therefore produces at least two distinct task-sequence signatures.

## 16. What prevents a task from running before its dependencies?

Plan construction verifies that each task depends only on earlier known task identifiers. During execution, the controller checks that every dependency is present in the completed-task set before dispatch. A violation moves the mission to FAILED with reason DEPENDENCY_NOT_SATISFIED and preserves the audit trail rather than silently changing the planned order.

## 17. How are budgets applied?

The mission context inherits explicit ceilings for steps, tool calls, cost units, artifacts, and retries per task. The complete plan must fit within the cost and call budget before execution begins. Each task then reserves and consumes its declared resources. Recovery also consumes resources. If the remaining budget cannot support the next action, the mission stops or aborts.

## 18. Why are budgets governance controls rather than only cost controls?

Budgets limit how long the system can continue, how many capabilities it can invoke, how often it can retry, and how many artifacts it can create. They reduce runaway loops and prevent an agent from converting uncertainty into unlimited activity. Increasing a budget is therefore an authority change that requires a new approved context, not a decision the active mission can make for itself.

## 19. What does the state machine control?

It controls movement among CREATED, VALIDATING, PLANNING, READY, RUNNING, PAUSED, RECOVERING, ESCALATED, COMPLETED, STOPPED, ABORTED, and FAILED. Every transition must appear in the declared allowlist. Invalid movements are denied and audited. Terminal missions cannot resume, which is directly tested by attempting to move a completed context back into execution.

## 20. What is recorded in the audit chain?

Each event records its sequence, mission identifier, current state, action, hash of the event payload, and previous event hash. The event then receives its own hash. Verification reconstructs every event body, confirms sequence continuity, checks the predecessor link, and recomputes the digest. This detects modification inside the notebook model, though production immutability needs protected storage and signatures.

## 21. What is contained in a checkpoint?

A checkpoint binds the mission identifier, state, step, remaining budget, hashes of all artifacts created so far, the current audit head, and a recoverability flag. Its checkpoint hash covers the complete record. The adversarial suite changes one nested budget value and confirms that verification fails, illustrating why a checkpoint must protect state as well as file identity.

## 22. How does schema validation work?

Every task label maps to a declared output type and required field set. The validator confirms the type, checks for missing fields, and rejects an output that attempts to assert live authority. A task is complete only after the output passes this contract. Typed validation prevents a plausible sentence from being accepted as the wrong institutional artifact.

## 23. How does recovery work?

The recovery mission injects a schema failure at strategy evaluation. The system records the task failure, enters RECOVERING, charges a recovery cost, writes a checkpoint, verifies that the retry allowance remains, and returns to RUNNING. The intentionally incomplete output is rejected. A corrected output then passes the ordinary validator and the mission continues. Only one retry is allowed.

## 24. Does recovery erase the original failure?

No. The failure event, recovery transition, rejected-output event, checkpoint, resource consumption, and final corrected artifact all remain in the record. The final mission carries a recovered flag, and the monitor counts recovered exceptions. Recovery is a successful governance outcome only because the exception remains observable and the authorized remediation contract was followed.

They demonstrate single-sector research, multi-sector and multi-regime adaptation, portfolio construction with independent risk, deterministic schema-failure recovery, and human escalation for promotion. ## Conclusion: bounded research autonomy and the NB07 bridge
