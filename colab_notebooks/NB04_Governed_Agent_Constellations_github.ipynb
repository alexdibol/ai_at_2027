# NB04 — Governed Agent Constellations

## Introduction

NB04 moves the course from bounded individual agency to governed institutional coordination. The preceding notebooks constructed the layers required for this transition. NB00A stated the operating protocol for an autonomous research system. NB00 created a synthetic equity market in which every later experiment could be reproduced without claiming access to live markets. NB00B demonstrated the complete research chain through a minimal KNN vertical slice. NB01 expanded that chain into a laboratory of prediction models, stock-level signals, strategies, portfolios, and comparative evaluation. NB02 placed research outputs behind execution simulation, independent risk, governance, provenance, and audit controls. NB03 then converted validated functions into 35 tools, nine skills, and 13 bounded agent roles. NB04 now asks what happens when those roles must work together.

The central object of this notebook is the governed agent constellation. A constellation is not an informal collection of agents and it is not a larger agent created by pooling permissions. It is a versioned institutional team whose objective, membership, admission policy, shared state, coordination rules, resource budget, handoff contract, approval checkpoints, prohibited actions, and validation status are explicitly declared. This distinction is essential. Individual agents can be well specified while their interactions remain unsafe. Coordination introduces new failure modes: circular delegation, deadlock, inconsistent assumptions, contradictory evidence, silent changes to shared state, resource exhaustion, role substitution, and the temptation to interpret majority agreement as authority. The constellation layer is designed to make those risks visible and testable.

The notebook creates four constellations that mirror the division of labor of an investment institution. The Research Team combines the Research Analyst, Data Scientist, and Experiment Designer to produce point-in-time, evidence-backed findings. The Strategy Team combines the Strategy Designer, Execution Simulation Agent, and Governance Strategy Evaluator to translate validated findings into bounded proposals and simulated implementation evidence. The Portfolio Team combines the Portfolio Constructor, Risk Officer, and Execution Simulation Agent to build and challenge a constrained synthetic-equity portfolio. The Risk Governance Team combines the Risk Officer, Governance Critic, Audit Reviewer, and Governance Strategy Evaluator to provide independent challenge, preserve dissent, verify evidence, and audit material recommendations. Membership overlaps where functional participation is necessary, but accountability does not disappear into the group.

Admission is deliberately strict. Only agents already registered and validated in NB03 may participate. Every admitted role must continue to deny live execution and self-approval. Joining a team does not enlarge an agent’s skill or tool allowlist, and the union of team capabilities does not become the authority of every member. Each delegated action remains subject to the effective-authority intersection established in NB03: system policy, agent permissions, skill allowlist, tool contract, and request scope must all permit the action. In practical terms, a Risk Officer’s presence in a portfolio constellation does not give a Portfolio Constructor the authority to override risk, and an Execution Simulation Agent’s technical knowledge does not create broker connectivity or live-trading permission.

Shared state is therefore treated as a governed record rather than a communal scratchpad. Twelve consequential fields receive explicit owners. The constellation coordinator owns the mission identifier, open questions, and budget state. The human governance owner owns the objective and approval state. The originating agent owns its assumptions; the producing agent owns the evidence it creates; the Audit Reviewer owns evidence hashes and the audit head; the Governance Critic owns the disagreement record; and the Risk Officer owns the risk state. Reviews append challenges and counterevidence instead of silently overwriting another role’s contribution. This ownership map makes it possible to reconstruct who stated an assumption, who produced evidence, who challenged it, who assessed risk, and who authorized a material decision.

Coordination is also encoded rather than improvised. Each constellation defines an ordered sequence, identifies work that may safely proceed in parallel, names a critic, describes advisory voting, and assigns unresolved ties to the human governance owner. Parallelism is permitted only where responsibilities are independent and shared-state ownership remains unambiguous. Voting can summarize professional views, but it cannot manufacture authorization. A unanimous group of research agents still cannot approve strategy promotion, waive a risk limit, or request deployment. The notebook therefore distinguishes coordination efficiency from decision rights: teams can accelerate analysis while material authority remains separated and accountable.

Dissent receives its own lifecycle. A disagreement may remain OPEN, move to HUMAN_REVIEW, become RESOLVED, or be retained as ACCEPTED_DISSENT. The last state matters because institutions sometimes act while a documented professional disagreement remains. Forced consensus would erase information precisely when governance needs it most. NB04 instead preserves the objection, the associated evidence, the accountable reviewer, and the decision pathway. Silence never counts as consent, contradictory evidence is not averaged away, and an unresolved conflict cannot be hidden by changing the shared assumptions without attribution.

Four human checkpoints establish the boundaries of autonomous coordination. CP1 governs mission admission when the objective is ambiguous or the mandate changes materially. CP2 governs promotion of a strategy beyond experimental status. CP3 governs any risk exception, override, or limit change. CP4 governs external-platform writing, connectivity, or deployment requests. These checkpoints do not require a human to perform every calculation. They reserve the moments at which the meaning, risk, status, or external effect of the work changes. The result is a practical form of human accountability: agents may research, simulate, criticize, and recommend, while material changes of authority return to an identified human owner.

The failure laboratory tests whether this constitution survives adverse coordination. Eight cases include one valid workflow and seven challenges: circular delegation, deadlock, inconsistent assumptions, conflicting evidence, tool-budget exhaustion, token-budget exhaustion, and agent failure. Expected outcomes include completion, escalation, blocking, termination, and controlled recovery. A failure test passes when the constellation responds according to policy, not when it forces the mission to continue. Every result records a reason, recovery instruction, absence of live authority, expected status, pass flag, and content hash. Thus, interruption and refusal become observable control outcomes rather than unexplained system breakdowns.

The first governed workflow evaluates whether sector-neutral 20-day momentum survives simulated costs in synthetic equities. The research mission completes, but automatic promotion is denied pending human review. This is the notebook’s most important demonstration. More agents, more coordination, and more evidence do not dissolve the research-only boundary. The constellation can produce a complete recommendation package while still lacking authority to transform that package into an approved strategy or a live order.

Within the course progression, NB04 is the organizational bridge between the capability factory and meta-level orchestration. NB03 defined what each role may do. NB04 defines how roles cooperate without privilege expansion, loss of provenance, or suppression of dissent. NB05 will introduce a governed meta-agent that discovers capabilities, decomposes missions, routes tasks, and arbitrates handoffs across these validated constellations. NB06 will assemble the release-candidate research system, and NB07 will create the controlled research-to-LEAN experiment bridge. The governing principle remains constant throughout: autonomy may expand through validated composition, but authority must remain explicit, bounded, reviewable, and human-accountable.

## Frequently Asked Questions — NB04

## 1. What is a governed agent constellation?

A governed constellation is a versioned team of bounded agents assembled for a specific institutional objective. It declares its members, admission rules, shared-state ownership, sequence, safe parallel work, critic, advisory voting rule, budgets, handoff requirements, approval checkpoints, prohibited actions, and validation status. It coordinates existing authority; it does not create new authority.

## 2. Why not use one powerful general-purpose agent?

A single agent would combine research, recommendation, risk review, approval, and potentially execution inside one decision-maker. That design weakens challenge and makes errors difficult to attribute. NB04 separates duties so specialized roles can disagree, verify one another’s evidence, and route material decisions to an accountable owner without allowing the producer of a proposal to approve it.

## 3. Which four constellations are created?

The Research Team develops evidence-backed findings; the Strategy Team converts validated research into bounded strategy proposals and implementation simulations; the Portfolio Team constructs and challenges constrained portfolios; and the Risk Governance Team independently reviews risk, governance, provenance, and audit evidence. Together they reproduce the core analytical and control functions of an investment institution in research-only form.

## 4. How are agents admitted to a constellation?

Admission requires a registered agent, validated status, an explicit denial of live execution, and an explicit denial of self-approval. All 13 participants come from the NB03 registry. A name invented during coordination, an unvalidated role, or a role lacking the required prohibitions would fail admission rather than being accepted for convenience.

## 5. Do permissions expand when agents join a team?

No. This is the central security property of the notebook. Each action remains bounded by the acting agent’s own permissions, its skill allowlist, the tool contract, system policy, and the request scope. The constellation can coordinate complementary capabilities, but one member cannot borrow another member’s authority. Registered capability is not equivalent to routed capability.

## 6. What is shared state, and why does it need ownership?

Shared state is the governed mission record used across roles. It contains the mission identifier, objective, assumptions, evidence, hashes, open questions, disagreements, risk state, budget state, approval state, decision, and audit head. Explicit ownership prevents silent rewriting and allows every consequential statement or change to be attributed to the role responsible for it.

## 7. Who owns the twelve shared-state fields?

Coordination fields belong to the constellation coordinator; objectives and approvals belong to the human governance owner; assumptions and evidence remain attributable to their originating and producing agents; evidence hashes and the audit head belong to the Audit Reviewer; disagreements belong to the Governance Critic; risk state belongs to the Risk Officer; and final decisions belong to the authorized decision owner.

## 8. How are sequential and parallel activities distinguished?

Each constellation specifies an ordered sequence and, where safe, a parallel group. Parallel work is permitted only when tasks are independent, the required inputs already exist, and no two agents silently compete to own the same field. Work that changes the meaning, risk, approval, or external effect of a mission stays ordered and subject to a checkpoint.

## 9. What does “majority advisory only” mean?

A vote can summarize the professional balance of views, but it cannot replace an authorization rule. Even a unanimous vote cannot promote an experimental strategy, waive a risk limit, expand a budget, or authorize external deployment. If the matter requires approval or remains deadlocked, the constellation routes it to the human governance owner with the evidence and dissent intact.

## 10. How is dissent represented?

Disagreements move through four explicit states: OPEN, HUMAN_REVIEW, RESOLVED, and ACCEPTED_DISSENT. Accepted dissent permits a decision to proceed while preserving a reasoned objection and its evidence. This avoids false consensus and gives later reviewers an accurate record of uncertainty, minority analysis, and the conditions under which the decision was made.

## 11. What are the four human checkpoints?

CP1 covers ambiguous mission admission or a material mandate change. CP2 covers strategy promotion beyond experimentation. CP3 covers a risk exception, override, or limit change. CP4 covers external-platform writing, connectivity, or deployment. These checkpoints reserve authority over changes in mandate, status, risk tolerance, and external effect while leaving routine bounded calculations to the agents.

## 12. How do the semantic and action gates operate at the constellation level?

The semantic gate asks whether the mission is sufficiently clear to admit: the objective, evidence boundary, asset universe, requested output, assumptions, and decision owner must be intelligible. The action gate is applied to every delegation. It asks whether the acting agent, routed skill, invoked tool, remaining budget, evidence, and research-only policy all permit the proposed step. Passing mission admission never creates blanket permission for later actions.

## 13. What budgets constrain each constellation?

Every constellation is limited to 24 steps, 36 tool calls, and one retry per agent. These are aggregate organizational budgets, not permission to bypass the smaller limits defined for individual roles. Tool- or token-budget exhaustion causes controlled termination with a checkpointed partial record; the team may not silently increase its allocation or continue through unlogged work.

## 14. What information must accompany a handoff?

A valid handoff contains the mission identifier, evidence, assumptions, open questions, and a provenance hash. It is rejected when evidence is missing, ownership is unresolved, or an unregistered capability is requested. The handoff therefore transfers both the analytical content and the institutional conditions needed for the receiving role to evaluate it responsibly.

## 15. Which actions remain prohibited?

Every constellation prohibits live trading, self-approval, risk override, circular delegation, silent assumption mutation, and unlogged recovery. These prohibitions apply even when a workflow is urgent, a majority agrees, or a model reports strong results. Coordination may improve analysis, but it cannot erase the constitutional boundaries inherited from NB03.

## 16. What does the failure laboratory test?

It evaluates a valid workflow plus circular delegation, deadlock, inconsistent assumptions, conflicting evidence, tool-budget exhaustion, token-budget exhaustion, and agent failure. The cases test organizational behavior rather than predictive performance. Each must reach its expected governed state and record the reason, recovery instruction, live-authority flag, pass status, and result hash.

## 17. How do BLOCKED, ESCALATED, TERMINATED, and RECOVERED differ?

BLOCKED means a policy conflict prevents progress until its cause is corrected. ESCALATED means accountable judgment is required, as with deadlock or conflicting evidence. TERMINATED means a hard resource boundary has been reached and work must stop. RECOVERED means the system followed an authorized, logged recovery procedure without expanding permissions or concealing the original failure.

## 20. Does the notebook prove that multi-agent systems are safe or production-ready?

No. It validates deterministic specifications and scenario outcomes inside a synthetic-equity, research-only environment. It does not establish resilience to hostile production conditions, model-provider failures, real-time data corruption, regulatory requirements, broker integration, or live market impact. Production claims require additional security, operational, legal, and empirical evidence.

## 23. What is the final lesson for the user?

Multi-agent intelligence becomes institutionally useful only when cooperation is governed as carefully as individual capability. A team must preserve role boundaries, ownership, evidence, dissent, budgets, and decision rights. More agents should create deeper analysis and stronger challenge, not a larger pool of unaccountable authority. The legitimate system result may be completion, escalation, blocking, termination, or controlled recovery.
