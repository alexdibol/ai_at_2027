Planning then creates a dynamic fan-out/fan-in graph. One data task precedes a separate training task for each selected model. A baseline task runs beside the model branch. A comparison task depends on every model and baseline result. Strategy generation depends on that comparison. The portfolio grid depends on the strategy catalog. A KNN-only mission, a two-model mission, and a five-model mission consequently produce different task counts, resource requirements, strategy counts, and backtest grids.

The execution layer invokes real scikit-learn estimators through registered factory functions. Every model uses the same features and chronological partitions. Models are first fitted on training data and assessed on validation data; they are then refitted on the combined training and validation sample for final test prediction. Classification and regression outputs are converted into a shared probability interface. Research selection remains distinct from institutional approval. The winner receives the status RESEARCH_CHAMPION_CANDIDATE. Independent risk measures gross exposure, maximum single-name exposure, maximum sector exposure, and stressed drawdown against declared limits. Even when the controls pass, the report preserves dissent: one deterministic synthetic world cannot establish economic persistence. Promotion evidence remains insufficient, live authority remains false, and promotion remains denied pending human review.

NB08 therefore represents a stronger form of bounded autonomy than NB06 alone. The orchestrator does not merely select a department and receive a placeholder artifact. Those branches reunite in a standardized comparison, then feed model-dependent and rule-based strategies into a portfolio and cost-stress grid. A predeclared rule nominates a research candidate, after which independent risk and audit retain the right to challenge, deny promotion, and preserve unresolved questions.

## 2. Deterministic multi-sector, multi-regime market

The substrate contains 30 synthetic equities distributed across five sectors and observed for 720 business days. Calm, trend, crisis, and recovery regimes alter drift, volatility, volume, and spreads. A fixed seed makes the world reproducible. The output manifest records the complete scope and a content hash so every later experiment can identify the dataset it actually used.

```python
SECTORS = ["Technology", "Financials", "Healthcare", "Industrials", "Consumer"]
REGIME_PARAMETERS = {
    "calm": {"drift": 0.0002, "vol": 0.006, "spread": 5.0, "volume": 1.00},
    "trend": {"drift": 0.0007, "vol": 0.008, "spread": 6.0, "volume": 1.10},
    "crisis": {"drift": -0.0016, "vol": 0.020, "spread": 18.0, "volume": 1.70},
    "recovery": {"drift": 0.0011, "vol": 0.012, "spread": 9.0, "volume": 1.35},
}


def regime_for_day(index, total_days):
    fraction = index / total_days
    if fraction < 0.28:
        return "calm"
    if fraction < 0.52:
        return "trend"
    if fraction < 0.67:
        return "crisis"
    if fraction < 0.82:
        return "recovery"
    return "calm"


def generate_synthetic_equity_market(seed=42, n_assets=30, n_days=720):
    rng = np.random.default_rng(seed)
    dates = pd.bdate_range("2022-01-03", periods=n_days)
    sectors = np.array([SECTORS[index % len(SECTORS)] for index in range(n_assets)])
    tickers = np.array([f"SYN{index + 1:02d}" for index in range(n_assets)])
    returns = np.zeros((n_days, n_assets))
    volume = np.zeros_like(returns)
    spread = np.zeros_like(returns)
    regimes = []
    asset_alpha = rng.normal(0, 0.00010, n_assets)

    for day in range(n_days):
        regime = regime_for_day(day, n_days)
        regimes.append(regime)
        params = REGIME_PARAMETERS[regime]
        market = rng.normal(params["drift"], params["vol"])
        sector_factor = {sector: rng.normal(0, params["vol"] * 0.35) for sector in SECTORS}
        previous = returns[day - 1] if day else np.zeros(n_assets)
        for asset in range(n_assets):
            continuation = 0.12 * previous[asset] if regime in {"trend", "recovery"} else -0.05 * previous[asset]
            shock = rng.normal(0, params["vol"] * 0.75)
            returns[day, asset] = market + sector_factor[sectors[asset]] + continuation + asset_alpha[asset] + shock
            volume[day, asset] = rng.lognormal(13.2, 0.35) * params["volume"]
            spread[day, asset] = max(1.0, params["spread"] * rng.lognormal(0, 0.18))

    prices = 100 * np.exp(np.cumsum(np.clip(returns, -0.20, 0.20), axis=0))
    records = []
    for day, date in enumerate(dates):
        for asset in range(n_assets):
            records.append({
                "date": date,
                "instrument_id": asset + 1,
                "ticker": tickers[asset],
                "sector": sectors[asset],
                "regime": regimes[day],
                "close": float(prices[day, asset]),
                "volume": int(volume[day, asset]),
                "spread_bps": float(spread[day, asset]),
            })
    frame = pd.DataFrame(records).sort_values(["instrument_id", "date"]).reset_index(drop=True)
    manifest = {
        "dataset_id": "NB08_SYNTHETIC_EQUITIES_V1",
        "seed": seed,
        "assets": n_assets,
        "days": n_days,
        "rows": len(frame),
        "sectors": sorted(frame["sector"].unique()),
        "regimes": sorted(frame["regime"].unique()),
        "content_hash": stable_hash(frame.round(8).to_dict("records")),
    }
    return frame, manifest


MARKET, DATASET_MANIFEST = generate_synthetic_equity_market()
assert len(MARKET) == 30 * 720
assert set(MARKET["sector"]) == set(SECTORS)
assert set(MARKET["regime"]) == set(REGIME_PARAMETERS)
print("deterministic multi-sector, multi-regime synthetic-equity substrate PASS")
```

## 3. Point-in-time features and chronological evaluation

NB08 reconstructs the six-feature NB01 research interface and creates a next-day target without allowing future observations into current features. The chronological splitter produces non-overlapping training, validation, and test periods. This common evidence contract is essential because model comparison is meaningful only when every candidate sees the same information under the same timing rules.

```python
FEATURES = ["return_1", "momentum_5", "momentum_20", "volatility_20", "volume_z20", "spread_bps"]


def build_point_in_time_features(market):
    data = market.sort_values(["instrument_id", "date"]).copy()
    grouped = data.groupby("instrument_id", group_keys=False)
    data["return_1"] = grouped["close"].pct_change()
    data["momentum_5"] = grouped["close"].pct_change(5)
    data["momentum_20"] = grouped["close"].pct_change(20)
    data["volatility_20"] = grouped["return_1"].transform(lambda values: values.rolling(20).std())
    log_volume = np.log1p(data["volume"])
    rolling_mean = log_volume.groupby(data["instrument_id"]).transform(lambda values: values.rolling(20).mean())
    rolling_std = log_volume.groupby(data["instrument_id"]).transform(lambda values: values.rolling(20).std())
    data["volume_z20"] = (log_volume - rolling_mean) / rolling_std
    data["forward_return_1"] = grouped["close"].pct_change().shift(-1)
    data["label_up"] = (data["forward_return_1"] > 0).astype(int)
    return data.dropna(subset=FEATURES + ["forward_return_1"]).reset_index(drop=True)


def chronological_split(data):
    dates = np.array(sorted(data["date"].unique()))
    train_end = int(0.60 * len(dates))
    validation_end = int(0.80 * len(dates))
    split = {
        "train": data[data["date"].isin(dates[:train_end])].copy().reset_index(drop=True),
        "validation": data[data["date"].isin(dates[train_end:validation_end])].copy().reset_index(drop=True),
        "test": data[data["date"].isin(dates[validation_end:])].copy().reset_index(drop=True),
    }
    assert split["train"]["date"].max() < split["validation"]["date"].min()
    assert split["validation"]["date"].max() < split["test"]["date"].min()
    return split


FEATURE_DATA = build_point_in_time_features(MARKET)
SPLIT = chronological_split(FEATURE_DATA)
FEATURE_MANIFEST = {
    "feature_names": FEATURES,
    "rows": len(FEATURE_DATA),
    "train_rows": len(SPLIT["train"]),
    "validation_rows": len(SPLIT["validation"]),
    "test_rows": len(SPLIT["test"]),
    "point_in_time": True,
    "chronological_non_overlap": True,
    "content_hash": stable_hash(FEATURE_DATA.round(8).to_dict("records")),
}
assert np.isfinite(FEATURE_DATA[FEATURES].to_numpy()).all()
print("point-in-time features and chronological train-validation-test partitions PASS")
```

```python
def logistic_factory(seed):
    return Pipeline([("scale", StandardScaler()), ("model", LogisticRegression(max_iter=1000, random_state=seed))])


def knn_factory(seed):
    return Pipeline([("scale", StandardScaler()), ("model", KNeighborsClassifier(n_neighbors=25, weights="distance"))])


def random_forest_factory(seed):
    return RandomForestClassifier(n_estimators=80, max_depth=5, min_samples_leaf=20, random_state=seed, n_jobs=1)


def neural_network_factory(seed):
    return Pipeline([
        ("scale", StandardScaler()),
        ("model", MLPClassifier(hidden_layer_sizes=(12,), max_iter=140, early_stopping=True, random_state=seed)),
    ])


def linear_regression_factory(seed):
    return Pipeline([("scale", StandardScaler()), ("model", LinearRegression())])


MODEL_FACTORIES = {
    "logistic_regression_trainer": logistic_factory,
    "knn_model_trainer": knn_factory,
    "random_forest_trainer": random_forest_factory,
    "bounded_neural_network_trainer": neural_network_factory,
    "linear_regression_trainer": linear_regression_factory,
}

MODEL_REGISTRY = {
    "logistic_regression_trainer": {"model_id": "logistic", "kind": "classifier", "family": "linear", "status": "validated", "scope": "synthetic_equities", "live_authority": False},
    "knn_model_trainer": {"model_id": "knn", "kind": "classifier", "family": "instance_based", "status": "validated", "scope": "synthetic_equities", "live_authority": False},
    "random_forest_trainer": {"model_id": "random_forest", "kind": "classifier", "family": "ensemble", "status": "validated", "scope": "synthetic_equities", "live_authority": False},
    "bounded_neural_network_trainer": {"model_id": "bounded_mlp", "kind": "classifier", "family": "neural", "status": "validated", "scope": "synthetic_equities", "live_authority": False},
    "linear_regression_trainer": {"model_id": "linear_regression", "kind": "regressor", "family": "linear", "status": "validated", "scope": "synthetic_equities", "live_authority": False},
}

BASELINE_REGISTRY = {
    "naive_up": {"kind": "benchmark", "champion_eligible": False},
    "random_seeded": {"kind": "benchmark", "champion_eligible": False},
    "moving_average": {"kind": "benchmark", "champion_eligible": False},
    "buy_and_hold": {"kind": "benchmark", "champion_eligible": False},
    "ar1_time_series": {"kind": "benchmark", "champion_eligible": False},
}

STRATEGY_REGISTRY = {
    "directional_probability": {"requires_model": True, "status": "validated", "scope": "synthetic_equities"},
    "trend_momentum": {"requires_model": False, "status": "validated", "scope": "synthetic_equities"},
    "cross_sectional_momentum": {"requires_model": False, "status": "validated", "scope": "synthetic_equities"},
    "mean_reversion": {"requires_model": False, "status": "validated", "scope": "synthetic_equities"},
    "regime_aware": {"requires_model": False, "status": "validated", "scope": "synthetic_equities"},
}

PORTFOLIO_REGISTRY = {
    "equal_weight": {"status": "validated"},
    "inverse_vol": {"status": "validated"},
    "vol_target": {"status": "validated"},
    "robust_mean_variance": {"status": "validated"},
}

TOOL_REGISTRY = {
    **{tool_id: {**spec, "capability": "model_development", "callable": True} for tool_id, spec in MODEL_REGISTRY.items()},
    "point_in_time_feature_builder": {"capability": "data_research", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
    "baseline_suite": {"capability": "model_development", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
    "model_comparison_engine": {"capability": "model_comparison", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
    "strategy_factory": {"capability": "strategy_design", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
    "portfolio_backtest_grid": {"capability": "backtesting", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
    "research_champion_selector": {"capability": "research_selection", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
    "independent_risk_engine": {"capability": "risk_review", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
    "audit_bundle_builder": {"capability": "audit_bundle", "status": "validated", "scope": "synthetic_equities", "live_authority": False, "callable": True},
}

AGENT_REGISTRY = {
    "research_analyst": {"constellation": "research_team", "allowed_tools": {"point_in_time_feature_builder"}},
    "data_scientist": {"constellation": "research_team", "allowed_tools": set(MODEL_REGISTRY) | {"baseline_suite"}},
    "experiment_designer": {"constellation": "research_team", "allowed_tools": {"model_comparison_engine", "portfolio_backtest_grid"}},
    "strategy_designer": {"constellation": "strategy_team", "allowed_tools": {"strategy_factory"}},
    "portfolio_constructor": {"constellation": "portfolio_team", "allowed_tools": {"research_champion_selector"}},
    "risk_officer": {"constellation": "risk_governance_team", "allowed_tools": {"independent_risk_engine"}},
    "audit_reviewer": {"constellation": "risk_governance_team", "allowed_tools": {"audit_bundle_builder"}},
}


def registry_snapshot():
    serializable_agents = {agent: {**spec, "allowed_tools": sorted(spec["allowed_tools"])} for agent, spec in AGENT_REGISTRY.items()}
    snapshot = {
        "models": MODEL_REGISTRY,
        "baselines": BASELINE_REGISTRY,
        "strategies": STRATEGY_REGISTRY,
        "portfolios": PORTFOLIO_REGISTRY,
        "tools": TOOL_REGISTRY,
        "agents": serializable_agents,
    }
    snapshot["registry_hash"] = stable_hash(snapshot)
    return snapshot


REGISTRY_SNAPSHOT = registry_snapshot()
assert set(MODEL_REGISTRY) == set(MODEL_FACTORIES)
assert all(spec["status"] == "validated" and not spec["live_authority"] for spec in TOOL_REGISTRY.values())
assert all(any(tool_id in agent["allowed_tools"] for agent in AGENT_REGISTRY.values()) for tool_id in TOOL_REGISTRY)
print("executable model, strategy, portfolio, tool, and agent registries PASS")
```

## 5. Objective interpretation, candidate discovery, and dynamic planning

An admitted objective may specify models explicitly, name them in natural language, or request a diversified comparison. Discovery normalizes aliases, denies unknown models, checks scope and authority, and binds the result to the registry hash. Planning fans out one training task per selected model, reunites the evidence in a comparison task, and then routes strategy, backtesting, selection, independent risk, and audit through least-privileged agent allowlists.

```python
REQUIRED_MISSION_FIELDS = {"mission_id", "objective", "asset_class", "scope", "evidence", "requested_output", "decision_owner"}
MODEL_ALIASES = {
    "logistic": "logistic_regression_trainer",
    "logistic regression": "logistic_regression_trainer",
    "knn": "knn_model_trainer",
    "k-nearest neighbors": "knn_model_trainer",
    "random forest": "random_forest_trainer",
    "neural": "bounded_neural_network_trainer",
    "neural network": "bounded_neural_network_trainer",
    "mlp": "bounded_neural_network_trainer",
    "linear regression": "linear_regression_trainer",
}
MATERIAL_OUTPUTS = {"strategy_promotion", "risk_override", "deployment", "live_execution"}


def interpret_mission(spec):
    missing = sorted(REQUIRED_MISSION_FIELDS - set(spec))
    if missing:
        return {"decision": "ESCALATE", "reason": "SEMANTIC_INSUFFICIENCY", "missing": missing}
    text = " ".join([spec["objective"], spec["scope"], " ".join(spec.get("side_effects", []))]).lower()
    if spec["asset_class"] != "synthetic_equities":
        return {"decision": "DENY", "reason": "NON_SYNTHETIC_SCOPE"}
    if any(term in text for term in ["live order", "send order", "broker", "deploy to production"]):
        return {"decision": "DENY", "reason": "LIVE_AUTHORITY_PROHIBITED"}
    if spec["requested_output"] in MATERIAL_OUTPUTS or "promote" in text:
        return {"decision": "REQUIRE_HUMAN_APPROVAL", "reason": "MATERIAL_DECISION"}
    parsed = {**deepcopy(spec), "normalized_objective": re.sub(r"\s+", " ", spec["objective"].strip().lower())}
    parsed["semantic_hash"] = stable_hash(parsed)
    return {"decision": "ADMIT", "reason": "MISSION_WITHIN_BOUNDARY", "parsed": parsed}


def discover_model_tools(parsed):
    requested = parsed.get("requested_models")
    objective = parsed["normalized_objective"]
    selected = []
    rationale = ""
    if requested:
        for name in requested:
            normalized = str(name).strip().lower()
            tool_id = MODEL_ALIASES.get(normalized, normalized)
            if tool_id not in MODEL_REGISTRY:
                return {"decision": "DENY", "reason": "UNKNOWN_MODEL", "unknown_model": name}
            selected.append(tool_id)
        rationale = "EXPLICIT_CANDIDATE_SET"
    else:
        mentioned = [tool_id for alias, tool_id in MODEL_ALIASES.items() if alias in objective]
        selected = list(dict.fromkeys(mentioned))
        if selected and not any(term in objective for term in ["compare", "best", "champion", "all models"]):
            rationale = "OBJECTIVE_NAMED_MODELS"
        else:
            selected = list(MODEL_REGISTRY)
            rationale = "DIVERSIFIED_COMPARATIVE_POLICY"

    eligible = []
    for tool_id in selected:
        spec = TOOL_REGISTRY[tool_id]
        if spec["status"] == "validated" and spec["scope"] == parsed["asset_class"] and not spec["live_authority"]:
            eligible.append(tool_id)
    if not eligible:
        return {"decision": "DENY", "reason": "NO_ELIGIBLE_MODEL_TOOLS"}
    result = {"decision": "DISCOVERED", "selected_tools": eligible, "rationale": rationale}
    result["discovery_hash"] = stable_hash({"registry_hash": REGISTRY_SNAPSHOT["registry_hash"], **result})
    return result


def select_agent_for_tool(tool_id, producer_agents=None, independence_required=False):
    producer_agents = set(producer_agents or [])
    candidates = []
    considered = {}
    for agent_id, spec in AGENT_REGISTRY.items():
        reasons = []
        if tool_id not in spec["allowed_tools"]:
            reasons.append("TOOL_NOT_ALLOWLISTED")
        if independence_required and agent_id in producer_agents:
            reasons.append("INSUFFICIENT_INDEPENDENCE")
        considered[agent_id] = "ELIGIBLE" if not reasons else ",".join(reasons)
        if not reasons:
            candidates.append((len(spec["allowed_tools"]), agent_id))
    if not candidates:
        return {"decision": "DENY", "reason": "NO_AUTHORIZED_AGENT", "considered": considered}
    candidates.sort()
    return {"decision": "ROUTE", "agent": candidates[0][1], "reason": "LEAST_PRIVILEGE_ALLOWLIST_MATCH", "considered": considered}


def build_dynamic_plan(parsed, discovery):
    tasks = []
    producer_agents = set()

    def add(description, tool_id, depends_on, cost_units, independence_required=False):
        route = select_agent_for_tool(tool_id, producer_agents, independence_required)
        if route["decision"] != "ROUTE":
            raise ValueError(route["reason"])
        task = {
            "task_id": f"T{len(tasks) + 1:02d}",
            "description": description,
            "tool_id": tool_id,
            "depends_on": list(depends_on),
            "cost_units": cost_units,
            "owner": route["agent"],
            "constellation": AGENT_REGISTRY[route["agent"]]["constellation"],
            "route_reason": route["reason"],
            "independence_required": independence_required,
        }
        task["task_hash"] = stable_hash(task)
        tasks.append(task)
        if not independence_required:
            producer_agents.add(route["agent"])
        return task["task_id"]

    data_id = add("Build validated point-in-time features and chronological partitions.", "point_in_time_feature_builder", [], 5)
    model_ids = [
        add(f"Train and evaluate {MODEL_REGISTRY[tool_id]['model_id']}.", tool_id, [data_id], 10)
        for tool_id in discovery["selected_tools"]
    ]
    baseline_id = add("Generate five non-promotable benchmark prediction streams.", "baseline_suite", [data_id], 5)
    compare_id = add("Compare validation and test diagnostics across trained models and benchmarks.", "model_comparison_engine", model_ids + [baseline_id], 5)
    strategy_id = add("Generate model-dependent and rule-based strategy signals.", "strategy_factory", [compare_id], 5)
    expected_strategies = len(model_ids) + 4
    expected_trials = expected_strategies * len(PORTFOLIO_REGISTRY)
    backtest_id = add("Run the complete strategy-portfolio grid under three cost assumptions.", "portfolio_backtest_grid", [strategy_id], expected_trials)
    champion_id = add("Select a research champion using the predeclared robust scorecard.", "research_champion_selector", [backtest_id], 3)
    risk_id = add("Independently challenge constraints, stress behavior, and promotion evidence.", "independent_risk_engine", [champion_id], 5, True)
    add("Assemble the provenance-bearing audit and handoff bundle.", "audit_bundle_builder", [risk_id], 3, True)

    known = set()
    for task in tasks:
        if not set(task["depends_on"]).issubset(known):
            raise ValueError("INVALID_DEPENDENCY_GRAPH")
        known.add(task["task_id"])
    plan = {
        "mission_id": parsed["mission_id"],
        "model_discovery": discovery,
        "tasks": tasks,
        "selected_model_tools": discovery["selected_tools"],
        "expected_strategies": expected_strategies,
        "expected_strategy_portfolio_trials": expected_trials,
        "total_cost_units": sum(task["cost_units"] for task in tasks),
    }
    plan["plan_hash"] = stable_hash(plan)
    return plan


PLAN_DEMO_ALL = build_dynamic_plan(interpret_mission({
    "mission_id": "DEMO_ALL", "objective": "Compare all models and select the most robust research champion.",
    "asset_class": "synthetic_equities", "scope": "multi-sector and multi-regime research",
    "evidence": ["NB00A-NB06"], "requested_output": "research_champion", "decision_owner": "human_governance_owner",
})["parsed"], {"decision": "DISCOVERED", "selected_tools": list(MODEL_REGISTRY), "rationale": "DIVERSIFIED_COMPARATIVE_POLICY", "discovery_hash": "demo"})
assert len(PLAN_DEMO_ALL["selected_model_tools"]) == 5
assert PLAN_DEMO_ALL["expected_strategy_portfolio_trials"] == 36
print("objective interpretation, model discovery, dynamic fan-out plan, and agent routing PASS")
```

## 6. Real model invocation and standardized diagnostics

Each selected model tool fits a real estimator. Validation metrics are produced before the final refit on combined training and validation data. The test interface standardizes classifier and regressor outputs as probabilities and records accuracy, balanced accuracy, Brier score, dispersion, confidence, prediction provenance, model family, tool identity, and champion eligibility. Five non-promotable baselines provide the comparison floor.

```python
def model_probabilities(model, features, kind, reference_std):
    if hasattr(model, "predict_proba"):
        return model.predict_proba(features)[:, 1]
    prediction = model.predict(features)
    scale = 4 * reference_std if reference_std else 1.0
    return np.clip(0.5 + prediction / scale, 0, 1)


def score_predictions(y_true, probabilities):
    probabilities = np.asarray(probabilities, dtype=float)
    return {
        "accuracy": float(accuracy_score(y_true, probabilities >= 0.5)),
        "balanced_accuracy": float(balanced_accuracy_score(y_true, probabilities >= 0.5)),
        "brier": float(brier_score_loss(y_true, probabilities)),
        "probability_mean": float(probabilities.mean()),
        "probability_std": float(probabilities.std()),
        "high_confidence_share": float(((probabilities > 0.60) | (probabilities < 0.40)).mean()),
    }


def invoke_model_tool(tool_id, split, seed=42):
    if tool_id not in MODEL_REGISTRY or tool_id not in MODEL_FACTORIES:
        raise ValueError("UNREGISTERED_MODEL_TOOL")
    spec = MODEL_REGISTRY[tool_id]
    if spec["status"] != "validated" or spec["live_authority"]:
        raise PermissionError("MODEL_TOOL_NOT_AUTHORIZED")
    model = MODEL_FACTORIES[tool_id](seed)
    train = split["train"]
    validation = split["validation"]
    train_validation = pd.concat([train, validation], ignore_index=True)
    target_train = train["forward_return_1"] if spec["kind"] == "regressor" else train["label_up"]
    model.fit(train[FEATURES], target_train)
    validation_probability = model_probabilities(model, validation[FEATURES], spec["kind"], train["forward_return_1"].std())
    validation_metrics = score_predictions(validation["label_up"], validation_probability)

    refit = MODEL_FACTORIES[tool_id](seed)
    target_refit = train_validation["forward_return_1"] if spec["kind"] == "regressor" else train_validation["label_up"]
    refit.fit(train_validation[FEATURES], target_refit)
    test_probability = model_probabilities(refit, split["test"][FEATURES], spec["kind"], train_validation["forward_return_1"].std())
    test_metrics = score_predictions(split["test"]["label_up"], test_probability)
    record = {
        "tool_id": tool_id,
        "model": spec["model_id"],
        "family": spec["family"],
        "kind": spec["kind"],
        "validation": validation_metrics,
        "test": test_metrics,
        "prediction_hash": stable_hash(np.round(test_probability, 10).tolist()),
        "champion_eligible": True,
        "live_authority": False,
    }
    record["artifact_hash"] = stable_hash(record)
    return {"record": record, "test_probability": test_probability, "fitted_model": refit}


def run_baseline_suite(split, seed=42):
    train_validation = pd.concat([split["train"], split["validation"]], ignore_index=True)
    test = split["test"]
    rng = np.random.default_rng(seed)
    coefficients = np.polyfit(train_validation["return_1"], train_validation["forward_return_1"], 1)
    ar_prediction = coefficients[0] * test["return_1"] + coefficients[1]
    scale = 4 * train_validation["forward_return_1"].std()
    streams = {
        "naive_up": np.ones(len(test)) * 0.51,
        "random_seeded": rng.random(len(test)),
        "moving_average": np.where(test["momentum_20"] > 0, 0.60, 0.40),
        "buy_and_hold": np.ones(len(test)) * 0.55,
        "ar1_time_series": np.clip(0.5 + ar_prediction / scale, 0, 1),
    }
    records = []
    for model_id, probability in streams.items():
        record = {
            "tool_id": "baseline_suite",
            "model": model_id,
            "family": "benchmark",
            "kind": "benchmark",
            "validation": None,
            "test": score_predictions(test["label_up"], probability),
            "prediction_hash": stable_hash(np.round(probability, 10).tolist()),
            "champion_eligible": False,
            "live_authority": False,
        }
        record["artifact_hash"] = stable_hash(record)
        records.append(record)
    return {"records": records, "streams": streams}


def compare_models(model_runs, baseline_result):
    rows = []
    for result in model_runs.values():
        record = result["record"]
        rows.append({
            "model": record["model"], "tool_id": record["tool_id"], "family": record["family"],
            "validation_balanced_accuracy": record["validation"]["balanced_accuracy"],
            "validation_brier": record["validation"]["brier"],
            "test_balanced_accuracy": record["test"]["balanced_accuracy"],
            "test_brier": record["test"]["brier"],
            "champion_eligible": True,
        })
    for record in baseline_result["records"]:
        rows.append({
            "model": record["model"], "tool_id": record["tool_id"], "family": record["family"],
            "validation_balanced_accuracy": np.nan, "validation_brier": np.nan,
            "test_balanced_accuracy": record["test"]["balanced_accuracy"],
            "test_brier": record["test"]["brier"], "champion_eligible": False,
        })
    return pd.DataFrame(rows).sort_values(["champion_eligible", "test_balanced_accuracy"], ascending=[False, False]).reset_index(drop=True)


MODEL_RUN_SMOKE = invoke_model_tool("knn_model_trainer", SPLIT)
assert len(MODEL_RUN_SMOKE["test_probability"]) == len(SPLIT["test"])
assert MODEL_RUN_SMOKE["record"]["tool_id"] == "knn_model_trainer"
assert not MODEL_RUN_SMOKE["record"]["live_authority"]
print("real registered model invocation and standardized diagnostic contracts PASS")
```

## 7. Model-dependent strategies, portfolios, cost stress, and research selection

Every trained model creates a distinct directional strategy, while trend momentum, cross-sectional momentum, mean reversion, and regime awareness remain independent rule-based competitors. Four portfolio methods produce a complete grid. Each combination is evaluated at 10, 25, and 50 basis points. The predeclared selector chooses the strongest worst-cost Sharpe, using stressed drawdown and turnover only as ordered tie-breakers.

```python
def generate_strategy_frames(test, model_runs):
    base = test[["date", "instrument_id", "ticker", "sector", "regime", "forward_return_1", "volatility_20", "momentum_5", "momentum_20", "return_1"]].copy().reset_index(drop=True)
    strategies = {}
    for result in model_runs.values():
        model_id = result["record"]["model"]
        probability = result["test_probability"]
        signal = np.where(probability > 0.56, 1, np.where(probability < 0.44, -1, 0))
        strategies[f"directional_{model_id}"] = base.assign(signal=signal, source_model=model_id, strategy_family="directional_probability")
    strategies["trend_momentum"] = base.assign(signal=np.sign(base["momentum_20"]).astype(int), source_model="rule_based", strategy_family="trend_momentum")
    ranks = base.groupby("date")["momentum_20"].rank(pct=True)
    strategies["cross_sectional_momentum"] = base.assign(signal=np.where(ranks >= 0.8, 1, np.where(ranks <= 0.2, -1, 0)), source_model="rule_based", strategy_family="cross_sectional_momentum")
    z_score = base["return_1"] / (base["volatility_20"] + 1e-9)
    strategies["mean_reversion"] = base.assign(signal=np.where(z_score < -1, 1, np.where(z_score > 1, -1, 0)), source_model="rule_based", strategy_family="mean_reversion")
    regime_signal = np.where(
        base["regime"].isin(["trend", "recovery"]), np.sign(base["momentum_20"]),
        np.where(base["regime"].eq("crisis"), -np.sign(base["momentum_5"]), 0),
    ).astype(int)
    strategies["regime_aware"] = base.assign(signal=regime_signal, source_model="rule_based", strategy_family="regime_aware")
    return strategies


def portfolio_backtest(frame, method="equal_weight", cost_bps=10.0, target_vol=0.10, max_weight=0.10, max_gross=1.0, max_sector=0.35):
    data = frame.sort_values(["instrument_id", "date"]).copy()
    active = data["signal"] != 0
    if method == "inverse_vol":
        raw = np.where(active, data["signal"] / (data["volatility_20"] + 1e-9), 0)
    elif method == "robust_mean_variance":
        raw = np.where(active, data["signal"] / (data["volatility_20"] ** 2 + 1e-6), 0)
    else:
        raw = data["signal"].astype(float)
    data["raw"] = raw
    denominator = data["raw"].abs().groupby(data["date"]).transform("sum").replace(0, np.nan)
    data["weight"] = (data["raw"] / denominator * max_gross).fillna(0).clip(-max_weight, max_weight)
    sector_gross = data["weight"].abs().groupby([data["date"], data["sector"]]).transform("sum")
    mask = sector_gross > max_sector
    data.loc[mask, "weight"] *= max_sector / sector_gross[mask]
    if method == "vol_target":
        scale = (target_vol / np.sqrt(252) / (data["volatility_20"].groupby(data["date"]).transform("mean") + 1e-9)).clip(upper=1)
        data["weight"] *= scale
    data["prior_weight"] = data.groupby("instrument_id")["weight"].shift().fillna(0)
    data["turnover"] = (data["weight"] - data["prior_weight"]).abs()
    data["net"] = data["weight"] * data["forward_return_1"] - data["turnover"] * cost_bps / 10000
    daily = data.groupby("date").agg(
        net_return=("net", "sum"), turnover=("turnover", "sum"),
        gross=("weight", lambda values: values.abs().sum()),
        max_abs_weight=("weight", lambda values: values.abs().max()),
    ).reset_index()
    daily["equity"] = (1 + daily["net_return"]).cumprod()
    daily["drawdown"] = daily["equity"] / daily["equity"].cummax() - 1
    sector_max = data.groupby(["date", "sector"])["weight"].apply(lambda values: values.abs().sum()).groupby("date").max().max()
    annual_return = daily["net_return"].mean() * 252
    annual_volatility = daily["net_return"].std() * np.sqrt(252)
    metrics = {
        "annualized_return": float(annual_return),
        "annualized_volatility": float(annual_volatility),
        "sharpe": float(annual_return / annual_volatility if annual_volatility else 0),
        "max_drawdown": float(daily["drawdown"].min()),
        "turnover": float(daily["turnover"].mean()),
        "max_gross": float(daily["gross"].max()),
        "max_weight": float(daily["max_abs_weight"].max()),
        "max_sector": float(sector_max),
        "constraint_breaches": int(((daily["gross"] > max_gross + 1e-9) | (daily["max_abs_weight"] > max_weight + 1e-9)).sum()),
    }
    return data, daily, metrics


def run_strategy_portfolio_grid(strategies, cost_grid=(10.0, 25.0, 50.0)):
    rows = []
    cache = {}
    for strategy_id, frame in strategies.items():
        for portfolio_id in PORTFOLIO_REGISTRY:
            cost_metrics = {}
            for cost_bps in cost_grid:
                detail, daily, metrics = portfolio_backtest(frame, portfolio_id, cost_bps)
                cost_metrics[cost_bps] = metrics
                cache[(strategy_id, portfolio_id, cost_bps)] = (detail, daily)
            base = cost_metrics[cost_grid[0]]
            row = {
                "strategy": strategy_id,
                "strategy_family": frame["strategy_family"].iloc[0],
                "source_model": frame["source_model"].iloc[0],
                "portfolio": portfolio_id,
                "annualized_return": base["annualized_return"],
                "annualized_volatility": base["annualized_volatility"],
                "sharpe": base["sharpe"],
                "max_drawdown": base["max_drawdown"],
                "turnover": base["turnover"],
                "max_gross": base["max_gross"],
                "max_weight": base["max_weight"],
                "max_sector": base["max_sector"],
                "constraint_breaches": base["constraint_breaches"],
                "sharpe_cost_25": cost_metrics[25.0]["sharpe"],
                "sharpe_cost_50": cost_metrics[50.0]["sharpe"],
                "worst_cost_sharpe": min(metrics["sharpe"] for metrics in cost_metrics.values()),
                "worst_cost_drawdown": min(metrics["max_drawdown"] for metrics in cost_metrics.values()),
            }
            rows.append(row)
    return pd.DataFrame(rows), cache


def select_research_champion(scorecard):
    eligible = scorecard[
        (scorecard["constraint_breaches"] == 0)
        & (scorecard["annualized_volatility"] <= 0.30)
        & np.isfinite(scorecard["worst_cost_sharpe"])
    ].copy()
    if eligible.empty:
        return {"decision": "DENY", "reason": "NO_GOVERNANCE_ELIGIBLE_CANDIDATE"}
    ranked = eligible.sort_values(["worst_cost_sharpe", "worst_cost_drawdown", "turnover"], ascending=[False, False, True])
    champion = ranked.iloc[0].to_dict()
    return {
        "decision": "RESEARCH_CHAMPION_CANDIDATE",
        "selection_rule": "maximize worst-cost Sharpe; then drawdown; then minimize turnover",
        "candidate": champion,
        "promotion": "DENIED_PENDING_HUMAN_REVIEW",
        "live_authority": False,
        "selection_hash": stable_hash(champion),
    }


STRATEGY_SMOKE = generate_strategy_frames(SPLIT["test"], {"knn_model_trainer": MODEL_RUN_SMOKE})
GRID_SMOKE, _ = run_strategy_portfolio_grid(STRATEGY_SMOKE)
assert len(STRATEGY_SMOKE) == 5
assert len(GRID_SMOKE) == 20
assert select_research_champion(GRID_SMOKE)["decision"] == "RESEARCH_CHAMPION_CANDIDATE"
print("model-dependent strategies, momentum strategies, portfolios, cost stresses, and research selection PASS")
```

## 8. Integrated fan-out/fan-in mission execution

The mission runner connects admission, discovery, planning, routing, budgets, state transitions, tool invocation, recovery, artifacts, traces, risk, and audit. The principal mission invokes all five model tools, compares ten total model and benchmark records, generates nine strategies, evaluates 36 portfolio combinations, and completes as research-only. The printed scorecards make both predictive and economic results visible.

```python
STATES = ["CREATED", "VALIDATING", "PLANNING", "READY", "RUNNING", "RECOVERING", "ESCALATED", "COMPLETED", "STOPPED", "ABORTED", "FAILED"]
TRANSITIONS = {
    "CREATED": ["VALIDATING"], "VALIDATING": ["PLANNING", "ESCALATED", "ABORTED"],
    "PLANNING": ["READY", "ESCALATED", "ABORTED"], "READY": ["RUNNING", "STOPPED"],
    "RUNNING": ["RECOVERING", "COMPLETED", "STOPPED", "FAILED"],
    "RECOVERING": ["RUNNING", "FAILED", "STOPPED"], "ESCALATED": [], "COMPLETED": [],
    "STOPPED": [], "ABORTED": [], "FAILED": [],
}


def create_context(spec, max_cost_units=200):
    return {
        "spec": deepcopy(spec), "state": "CREATED", "audit": [], "trace": [],
        "artifacts": [], "completed_tasks": set(), "budget_initial": max_cost_units,
        "budget_remaining": max_cost_units, "retry_count": {}, "recovered": False,
    }


def transition(context, target, reason):
    current = context["state"]
    if target not in TRANSITIONS.get(current, []):
        append_audit(context, "TRANSITION_DENIED", {"from": current, "to": target, "reason": reason})
        return False
    context["state"] = target
    append_audit(context, "STATE_TRANSITION", {"from": current, "to": target, "reason": reason})
    return True


def consume_budget(context, units, action):
    if context["budget_remaining"] < units:
        return False
    context["budget_remaining"] -= units
    append_audit(context, "BUDGET_CONSUMED", {"action": action, "units": units, "remaining": context["budget_remaining"]})
    return True


def record_tool_call(context, task, output_summary):
    artifact = {
        "task_id": task["task_id"], "tool_id": task["tool_id"], "owner": task["owner"],
        "constellation": task["constellation"], "output_hash": stable_hash(output_summary),
        "parent_hashes": [item["artifact_hash"] for item in context["artifacts"]],
        "live_authority": False,
    }
    artifact["artifact_hash"] = stable_hash(artifact)
    context["artifacts"].append(artifact)
    context["completed_tasks"].add(task["task_id"])
    append_audit(context, "TOOL_COMPLETED", artifact)
    trace = {
        "step": len(context["trace"]) + 1, "task_id": task["task_id"], "tool_id": task["tool_id"],
        "owner": task["owner"], "constellation": task["constellation"], "outcome": "COMPLETE",
        "budget_remaining": context["budget_remaining"], "artifact_hash": artifact["artifact_hash"],
    }
    trace["trace_hash"] = stable_hash(trace)
    context["trace"].append(trace)
    return artifact


def independent_risk_review(champion, cache):
    candidate = champion["candidate"]
    key = (candidate["strategy"], candidate["portfolio"], 50.0)
    detail, daily = cache[key]
    metrics = {
        "max_gross": float(daily["gross"].max()),
        "max_single_name": float(daily["max_abs_weight"].max()),
        "max_sector": float(detail.groupby(["date", "sector"])["weight"].apply(lambda values: values.abs().sum()).max()),
        "maximum_drawdown_at_50_bps": float(daily["drawdown"].min()),
    }
    limits = {"max_gross": 1.0, "max_single_name": 0.10, "max_sector": 0.35, "maximum_drawdown_at_50_bps": 0.20}
    checks = {
        "max_gross": metrics["max_gross"] <= limits["max_gross"] + 1e-9,
        "max_single_name": metrics["max_single_name"] <= limits["max_single_name"] + 1e-9,
        "max_sector": metrics["max_sector"] <= limits["max_sector"] + 1e-9,
        "maximum_drawdown_at_50_bps": metrics["maximum_drawdown_at_50_bps"] >= -limits["maximum_drawdown_at_50_bps"],
    }
    return {
        "status": "PASS_CONTROLS" if all(checks.values()) else "BREACH",
        "metrics": metrics, "limits": limits, "checks": checks,
        "promotion_evidence": "INSUFFICIENT",
        "dissent": ["one deterministic synthetic world cannot establish economic persistence"],
        "independent_owner": "risk_officer", "promotion": "DENIED_PENDING_HUMAN_REVIEW",
        "live_authority": False,
    }


def find_task(plan, tool_id):
    return next(task for task in plan["tasks"] if task["tool_id"] == tool_id)


def terminal_result(context, status, reason, plan=None):
    return {
        "status": status, "reason": reason, "state": context["state"], "plan": plan,
        "audit": deepcopy(context["audit"]), "trace": deepcopy(context["trace"]),
        "artifacts": deepcopy(context["artifacts"]), "budget_initial": context["budget_initial"],
        "budget_remaining": context["budget_remaining"], "recovered": context["recovered"],
        "live_authority": False,
    }


def run_integrated_mission(spec, max_cost_units=200):
    context = create_context(spec, max_cost_units)
    transition(context, "VALIDATING", "MISSION_RECEIVED")
    admission = interpret_mission(spec)
    append_audit(context, "ADMISSION_DECISION", admission)
    if admission["decision"] == "ESCALATE" or admission["decision"] == "REQUIRE_HUMAN_APPROVAL":
        transition(context, "ESCALATED", admission["reason"])
        return terminal_result(context, "ESCALATED", admission["reason"])
    if admission["decision"] == "DENY":
        transition(context, "ABORTED", admission["reason"])
        return terminal_result(context, "DENIED", admission["reason"])

    transition(context, "PLANNING", "OBJECTIVE_ADMITTED")
    discovery = discover_model_tools(admission["parsed"])
    append_audit(context, "MODEL_DISCOVERY", discovery)
    if discovery["decision"] != "DISCOVERED":
        transition(context, "ABORTED", discovery["reason"])
        return terminal_result(context, "DENIED", discovery["reason"])
    plan = build_dynamic_plan(admission["parsed"], discovery)
    append_audit(context, "PLAN_CREATED", {"plan_hash": plan["plan_hash"], "tasks": len(plan["tasks"]), "cost": plan["total_cost_units"]})
    if plan["total_cost_units"] > context["budget_remaining"]:
        transition(context, "ABORTED", "BUDGET_EXHAUSTED_BEFORE_EXECUTION")
        return terminal_result(context, "TERMINATED", "BUDGET_EXHAUSTED_BEFORE_EXECUTION", plan)
    transition(context, "READY", "PLAN_VALIDATED")
    transition(context, "RUNNING", "EXECUTION_STARTED")

    data_task = find_task(plan, "point_in_time_feature_builder")
    consume_budget(context, data_task["cost_units"], data_task["tool_id"])
    record_tool_call(context, data_task, FEATURE_MANIFEST)

    model_runs = {}
    for tool_id in discovery["selected_tools"]:
        task = find_task(plan, tool_id)
        if not set(task["depends_on"]).issubset(context["completed_tasks"]):
            transition(context, "FAILED", "DEPENDENCY_NOT_SATISFIED")
            return terminal_result(context, "FAILED", "DEPENDENCY_NOT_SATISFIED", plan)
        if spec.get("inject_failure_model") == tool_id and not context["retry_count"].get(tool_id):
            context["retry_count"][tool_id] = 1
            append_audit(context, "MODEL_TOOL_FAILURE", {"tool_id": tool_id, "reason": "INJECTED_RECOVERABLE_FAILURE"})
            transition(context, "RECOVERING", "MODEL_TOOL_FAILURE")
            if not consume_budget(context, 2, "RECOVERY"):
                transition(context, "STOPPED", "RECOVERY_BUDGET_EXHAUSTED")
                return terminal_result(context, "TERMINATED", "RECOVERY_BUDGET_EXHAUSTED", plan)
            append_audit(context, "CHECKPOINT", {"recoverable": True, "failed_tool": tool_id, "retry": 1})
            transition(context, "RUNNING", "ONE_PERMITTED_RETRY")
            context["recovered"] = True
        if not consume_budget(context, task["cost_units"], tool_id):
            transition(context, "STOPPED", "BUDGET_EXHAUSTED")
            return terminal_result(context, "TERMINATED", "BUDGET_EXHAUSTED", plan)
        result = invoke_model_tool(tool_id, SPLIT)
        model_runs[tool_id] = result
        record_tool_call(context, task, result["record"])

    baseline_task = find_task(plan, "baseline_suite")
    consume_budget(context, baseline_task["cost_units"], baseline_task["tool_id"])
    baselines = run_baseline_suite(SPLIT)
    record_tool_call(context, baseline_task, baselines["records"])

    compare_task = find_task(plan, "model_comparison_engine")
    consume_budget(context, compare_task["cost_units"], compare_task["tool_id"])
    model_scorecard = compare_models(model_runs, baselines)
    record_tool_call(context, compare_task, model_scorecard.round(8).to_dict("records"))

    strategy_task = find_task(plan, "strategy_factory")
    consume_budget(context, strategy_task["cost_units"], strategy_task["tool_id"])
    strategies = generate_strategy_frames(SPLIT["test"], model_runs)
    record_tool_call(context, strategy_task, sorted(strategies))

    backtest_task = find_task(plan, "portfolio_backtest_grid")
    consume_budget(context, backtest_task["cost_units"], backtest_task["tool_id"])
    strategy_scorecard, cache = run_strategy_portfolio_grid(strategies)
    record_tool_call(context, backtest_task, strategy_scorecard.round(8).to_dict("records"))

    champion_task = find_task(plan, "research_champion_selector")
    consume_budget(context, champion_task["cost_units"], champion_task["tool_id"])
    champion = select_research_champion(strategy_scorecard)
    if champion["decision"] != "RESEARCH_CHAMPION_CANDIDATE":
        transition(context, "FAILED", champion["reason"])
        return terminal_result(context, "FAILED", champion["reason"], plan)
    record_tool_call(context, champion_task, champion)

    risk_task = find_task(plan, "independent_risk_engine")
    consume_budget(context, risk_task["cost_units"], risk_task["tool_id"])
    risk_report = independent_risk_review(champion, cache)
    record_tool_call(context, risk_task, risk_report)

    audit_task = find_task(plan, "audit_bundle_builder")
    consume_budget(context, audit_task["cost_units"], audit_task["tool_id"])
    audit_summary = {
        "registry_hash": REGISTRY_SNAPSHOT["registry_hash"], "plan_hash": plan["plan_hash"],
        "dataset_hash": DATASET_MANIFEST["content_hash"], "feature_hash": FEATURE_MANIFEST["content_hash"],
        "model_scorecard_hash": stable_hash(model_scorecard.round(8).to_dict("records")),
        "strategy_scorecard_hash": stable_hash(strategy_scorecard.round(8).to_dict("records")),
        "champion_hash": champion["selection_hash"], "risk_hash": stable_hash(risk_report),
        "open_questions": risk_report["dissent"], "promotion": "DENIED_PENDING_HUMAN_REVIEW",
        "live_authority": False,
    }
    record_tool_call(context, audit_task, audit_summary)
    transition(context, "COMPLETED", "RESEARCH_MISSION_COMPLETE")
    result = terminal_result(context, "COMPLETED_RESEARCH_ONLY", "ALL_RUNTIME_GATES_PASSED", plan)
    result.update({
        "model_scorecard": model_scorecard, "strategy_scorecard": strategy_scorecard,
        "champion": champion, "risk_report": risk_report, "audit_summary": audit_summary,
        "model_runs": {tool_id: output["record"] for tool_id, output in model_runs.items()},
        "strategies": sorted(strategies), "cache": cache,
    })
    return result


MAIN_SPEC = {
    "mission_id": "NB08_M01_FULL_LABORATORY",
    "objective": "Compare all validated predictive models and momentum strategies, construct cost-aware portfolios, and identify the most robust research champion.",
    "asset_class": "synthetic_equities",
    "scope": "multi-sector and multi-regime synthetic-equity research",
    "evidence": ["NB00A", "NB00", "NB00B", "NB01", "NB02", "NB03", "NB04", "NB05", "NB06"],
    "requested_output": "research_champion",
    "decision_owner": "human_governance_owner",
}

MAIN_RESULT = run_integrated_mission(MAIN_SPEC)
assert MAIN_RESULT["status"] == "COMPLETED_RESEARCH_ONLY"
assert len(MAIN_RESULT["model_runs"]) == 5
assert len(MAIN_RESULT["strategies"]) == 9
assert len(MAIN_RESULT["strategy_scorecard"]) == 36
assert verify_audit_chain(MAIN_RESULT["audit"])
MAIN_SUMMARY = {
    "mission_status": MAIN_RESULT["status"],
    "model_tools_invoked": len(MAIN_RESULT["model_runs"]),
    "models_and_benchmarks_compared": len(MAIN_RESULT["model_scorecard"]),
    "strategies_generated": len(MAIN_RESULT["strategies"]),
    "strategy_portfolio_trials": len(MAIN_RESULT["strategy_scorecard"]),
    "research_champion": {
        key: MAIN_RESULT["champion"]["candidate"][key]
        for key in ["strategy", "portfolio", "worst_cost_sharpe", "worst_cost_drawdown"]
    },
    "promotion": MAIN_RESULT["champion"]["promotion"],
    "live_authority": MAIN_RESULT["live_authority"],
}
print(json.dumps(MAIN_SUMMARY, indent=2, default=str))
print("\nMODEL SCORECARD")
print(MAIN_RESULT["model_scorecard"].round(4).to_string(index=False))
print("\nTOP FIVE STRATEGY–PORTFOLIO CANDIDATES")
print(MAIN_RESULT["strategy_scorecard"].sort_values("worst_cost_sharpe", ascending=False).head(5).round(4).to_string(index=False))
print("integrated fan-out training, fan-in comparison, strategy grid, risk, and audit mission PASS")
```

## 10. Adversarial validation and final acceptance

The final gate tests missing semantics, unknown models, live-action intent, promotion, budget exhaustion, bounded recovery, missing tool permission, registry tampering, audit tampering, and self-promotion. It also compares one-, two-, and five-model plans to prove that candidate selection changes the executable graph. Acceptance requires every quantitative, governance, provenance, and research-only condition to pass simultaneously.

# Frequently Asked Questions — NB08

## 1. What problem does NB08 solve?

It reconnects the quantitative laboratory with the autonomous control architecture. Earlier notebooks had real models and strategies, while later notebooks had dynamic planning and governance. NB08 makes the orchestrator invoke the registered quantitative implementations, compare their outputs, construct and stress portfolios, nominate a research candidate, and preserve independent risk and audit.

## 2. Was the earlier architecture autonomous before NB08?

NB06 had bounded control-plane autonomy. It could interpret, decompose, label, route, monitor, recover, and close. It did not invoke the NB01 model and strategy functions. NB08 extends autonomy into the research execution plane. This still remains bounded autonomy because policies, registries, budgets, scopes, and human decision rights constrain every action.

## 3. Which upstream notebooks are integrated?

The manifest identifies NB00A through NB06. Their contributions include the constitutional boundary, synthetic market, KNN vertical slice, model–strategy–portfolio laboratory, execution and risk controls, tool–skill–agent registries, governed constellations, meta-agent planning, and the NB06 state machine. The hashes preserve the exact upstream versions used for the integration claim.

## 4. Does NB08 literally execute every cell from every earlier notebook?

No. That would duplicate large datasets, outputs, and notebook-specific setup. NB08 integrates validated interfaces and implementations into one self-contained runtime. It adapts the actual NB01 analytical functions and records upstream hashes. This is architectural integration, not concatenation of every prior cell.

## 5. Which predictive models are executable tools?

Five model tools are executable: logistic regression, K-nearest neighbors, random forest, bounded multilayer perceptron, and linear regression. Each has a separate identity, family, prediction kind, factory, validated lifecycle state, synthetic-equity scope, agent allowlist, output record, and explicit absence of live authority.

## 6. Why is KNN included separately?

KNN was central to the early vertical slice but became stranded in NB03: its tool was registered without being exposed through a skill. NB08 restores it as a routable, executable model tool available to the Data Scientist under the same evidence and governance conditions as the other candidates.

## 7. What role do the five baselines play?

Naive upward probability, seeded random predictions, moving average, buy and hold, and an AR(1)-style proxy prevent the laboratory from evaluating models in a vacuum. They are diagnostic benchmarks, not promotion candidates. Their champion-eligibility flag is therefore false even when a baseline happens to rank highly on one metric.

## 8. How does the orchestrator select model candidates?

Explicit requested models have priority. Otherwise, model names in the objective can restrict the set. An objective asking to compare, find the best, select a champion, or examine all models triggers the diversified comparative policy and selects every validated candidate. Unknown names are denied, and every selected tool must pass status, scope, and authority checks.

## 9. Does the orchestrator select the winning model before training?

No. Discovery selects a candidate set, not a winner. Each candidate receives the same data, features, partitions, and metric contract. Their results are then compared. This distinction prevents the planner from converting an untested preference into an apparent empirical conclusion.

## 10. How does the plan change with the model request?

A separate training task is created for every selected model. Consequently, a KNN-only mission has one model branch, a neural-versus-logistic mission has two, and the principal comparison has five. Strategy counts, backtest counts, costs, dependencies, and plan hashes change with those selections.

## 11. What does fan-out/fan-in mean?

Fan-out means that one validated dataset feeds several independent model-training branches. Fan-in means that the branches must return standardized evidence before the comparison task can proceed. The pattern supports parallel research while preserving a single comparison contract and explicit dependencies.

## 12. How are tasks routed to agents?

The route selector looks for agents whose allowlists contain the requested tool. It rejects agents lacking that tool and rejects a producer when an independent task would create self-review. Among eligible agents, it selects the one with the narrowest allowlist. The selected agent and its constellation are recorded in the plan and trace.

## 13. Are the tools merely names in a dictionary?

No. Every model tool ID maps one-to-one to a callable factory that constructs a real scikit-learn estimator. The mission runner resolves the registered ID, verifies its policy state, fits the estimator, generates validation and test predictions, calculates diagnostics, hashes the prediction stream, and records the completed invocation.

## 14. Why are both validation and test periods used?

Validation evidence measures behavior before final evaluation and could support future hyperparameter or candidate decisions. The final model is then refitted on training plus validation and evaluated on the untouched test period. This protects the test interval from becoming an informal tuning set.

## 15. How are regression and classification models compared?

Classifiers provide upward probabilities directly. Linear-regression predictions are transformed into bounded probability-like scores using the dispersion of forward returns. Every candidate then receives the same directional and calibration diagnostics. The transformation is an educational common interface, not a claim that all probability estimates are perfectly calibrated.

## 16. Which diagnostics are recorded?

The scorecard includes accuracy, balanced accuracy, Brier score, mean predicted probability, probability dispersion, high-confidence share, validation and test results, model family, tool identity, prediction hash, and champion eligibility. Economic selection is handled separately because prediction quality does not automatically imply a profitable strategy.

## 17. Which strategies are tested?

Each trained model produces a directional-probability strategy. Four rule-based strategies are added: trend momentum, cross-sectional momentum, mean reversion, and regime awareness. With five models, this produces nine strategy streams. The architecture therefore compares model-derived and economically specified signals on the same portfolio substrate.

## 18. Which portfolio methods are tested?

The methods are equal weighting, inverse volatility, volatility targeting, and a robust mean–variance proxy. Every method applies gross-exposure, single-name, and sector constraints. Residual exposure remains cash rather than being automatically levered to force full investment.

## 19. Why are several transaction-cost assumptions used?

The base, moderate-stress, and severe-stress assumptions are 10, 25, and 50 basis points. A strategy that wins only under the cheapest implementation assumption is fragile. The champion rule therefore uses the worst Sharpe across the three cost conditions instead of selecting on the base result alone.

## 20. How is the research champion chosen?

Candidates with constraint breaches, excessive volatility, or non-finite results are excluded. The selector then maximizes worst-cost Sharpe. If necessary, it prefers the less severe stressed drawdown and then lower turnover. The ordering is declared before results are inspected and is preserved in the selection record.

## 21. Does research champion mean approved strategy?

No. It means only that the candidate ranked first under the declared experimental rule in this synthetic world. The promotion field remains DENIED_PENDING_HUMAN_REVIEW. Research comparison, independent risk, strategy promotion, deployment, and live execution are distinct institutional decisions.

## 22. What does independent risk review examine?

The Risk Officer measures maximum gross exposure, maximum single-name weight, maximum sector exposure, and drawdown under the severe cost assumption. These values are tested against declared limits. The report also preserves dissent and states that one deterministic world cannot establish economic persistence.

## 23. Why can risk controls pass while promotion remains denied?

A limit pass means the simulated portfolio stayed within selected exposure and drawdown controls. It does not prove persistent alpha, realistic capacity, operational security, legal compliance, or production readiness. Risk acceptance within the experiment is evidence, not authority.

## 24. How are budgets enforced?

Every plan task declares cost units. The complete plan must fit the mission budget before execution. Each tool invocation then consumes its authorized amount, and recovery consumes additional units. A budget failure terminates the mission rather than allowing the system to enlarge its own resources.

## 25. How does model failure recovery work?

An injected model failure enters the RECOVERING state, records the failure, charges recovery cost, and creates a checkpoint event. One retry is permitted. The ordinary registered implementation must then succeed. Recovery remains visible in the audit chain and cannot erase the original exception.

## 26. What provenance is attached to tool calls?

Each artifact records its task, tool, owner, constellation, output hash, parent artifact hashes, live-authority state, and artifact hash. Audit events form a hash-linked chain. The execution trace records ordering, budget state, route, outcome, and artifact identity.

## 27. Which artifacts are written?

The output contract contains the capability registry, dynamic plan, model scorecard, strategy–portfolio scorecard, selected daily portfolio, selected positions, independent risk report, execution trace, runtime monitor, and audit bundle. These files allow downstream work to consume the evidence without rerunning every internal step.

## 28. What does the monitor measure?

It reports mission status, model tools invoked, strategies generated, strategy–portfolio trials, trace events, budget utilization, recovery status, audit-chain integrity, promotion state, and live authority. The monitor is derived from execution records rather than copied from an expected status table.

## 29. Which adversarial cases are tested?

The suite tests missing semantics, an unknown model, live-order intent, strategy promotion, budget exhaustion, recoverable model failure, a tool removed from every agent allowlist, registry tampering, audit-chain tampering, and system self-promotion. All ten expected outcomes must pass before acceptance.

## 30. What has NB08 proved?

It proves that the notebook can dynamically choose candidate model tools, invoke real implementations, produce model-dependent strategies, execute a constrained and cost-stressed portfolio grid, nominate a research candidate, preserve independent challenge, and close with auditable evidence under deterministic rules.

## 31. What has NB08 not proved?

It has not proved real-market profitability, data-vendor readiness, production security, distributed reliability, legal compliance, execution capacity, stable alpha, or safe live adaptation. It uses one deterministic synthetic world and simplified transaction-cost assumptions. Its conclusions must remain inside that evidence boundary.

## 32. Is the architecture now fully autonomous?

It is an integrated bounded autonomous research architecture. It can choose and execute approved quantitative research paths without manual assignment of every model and strategy. It is not an autonomous trading institution. Promotion, deployment, external writes, broker connections, and live orders remain outside its authority.

## 33. What should the next stage add?

The next stage should connect these governed artifacts to an experiment registry and a controlled LEAN-compatible representation, preserve code and data hashes, reconcile local and platform results, and require explicit approval for external writes. Production engineering would additionally require durable state, authentication, secrets management, observability, incident response, and independent security validation.
